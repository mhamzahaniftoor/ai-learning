# Topic 14 | Type Conversion & Casting in Python

Converting a value from one data type to another, for example integer to string, string to integer, integer to float, or float to integer.

## Two kinds
- **Implicit conversion (type conversion):** Python does it **automatically** when it is safe. `5 + 2.5` gives `7.5`, because the int `5` becomes `5.0` first.
- **Explicit conversion (type casting):** **we** do it using a conversion function.

## Casting functions

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int(x)` | Integer | `int("25")`, `int(3.9)` | `25`, `3` (decimal part is **cut off**, not rounded) |
| `float(x)` | Float | `float("3.5")`, `float(5)` | `3.5`, `5.0` |
| `str(x)` | String | `str(100)`, `str(3.0)` | `"100"`, `"3.0"` |
| `bool(x)` | Boolean | `bool(0)`, `bool("hi")` | `False`, `True` |

## Why do we need it?
- `input()` always returns a **string**, so numbers must be converted before using them in calculations.
- `"Age: " + 34` fails. Both sides of `+` must be strings, so convert with `str(34)`.
- Data read from files, forms and APIs often arrives as text.

## The `input()` function
`input("message")` shows a message, waits for the user to type, and returns what they typed **as a string**.

## Rules
- `int("3.5")` fails. Convert in two steps: `int(float("3.5"))`.
- `int("abc")`, `int("")` and `float("1,5")` fail with `ValueError`.
- `int()` and `float()` accept spaces around the number (`int("  42  ")` works).
- `bool()` is `False` for `0`, `0.0`, `""`, `[]` and `None`. Everything else is `True`. So `bool("False")` is **True**, because the text isn't empty.

```python
age = "34"
print(age)                  # 34 (print doesn't show quotes)
print(type(age))            # <class 'str'>
print(int(age) + 1)         # 35

# String <-> number
print(int("25"), float("3.5"), str(100), str(3.0))   # 25 3.5 100 3.0

# int <-> float
print(float(5))             # 5.0
print(int(3.9), int(-3.9))  # 3 -3 (cuts the decimals, no rounding)
print(round(3.9), round(3.2))   # 4 3 (rounds)
print(int(float("3.5")))    # 3 (two steps: string -> float -> int)

# bool
print(bool(0), bool(1), bool(""), bool("hi"), bool([]), bool(None))
print(bool("False"))        # True (any non-empty text is True)
print(int(True), int(False), float(True))   # 1 0 1.0

# Implicit conversion (Python does it for us)
print(5 + 2.5, type(5 + 2.5))   # 7.5 <class 'float'>
print(True + 1)                 # 2
print(10 / 2)                   # 5.0 (division always gives a float)

# + needs two strings
print("Age: " + str(34))        # Age: 34
```

```python
age = input("Please write your age ")
print(age, type(age))       # always a string, even if you type 34

age = int(age)              # convert before doing maths
print(type(age), age + 1)

name = input("Name: ")
price = float(input("Price: "))   # convert straight away
qty = int(input("Qty: "))
print(f"{name}: Rs.{price * qty}")
```

```python
for expr in ['int("abc")', 'int("3.5")', 'int("")', '"Age: " + 34']:
    try:
        eval(expr)
    except Exception as e:
        print(f"{expr:14} -> {type(e).__name__}: {e}")


def get_age(text):
    try:
        return int(text)
    except ValueError:
        print(f"'{text}' is not a valid whole number")
        return None

for t in ["21", "abc", "3.5", ""]:
    print(repr(t), "->", get_age(t))
```

## Key Points
- **Type conversion** (implicit) is automatic. **Type casting** (explicit) uses `int()`, `float()`, `str()` and `bool()`.
- `input()` **always returns a string**. Convert it with `int()` or `float()` before maths.
- `int(3.9)` gives `3` (the decimal part is cut off). Use `round()` if you want rounding.
- `int("3.5")` fails. Use `int(float("3.5"))`.
- You can't join a string and a number with `+`. Use `str()` or an f-string.
- A string that doesn't look like a number gives a `ValueError`. Use `try/except` to handle it.
- `bool()` is `False` only for empty or zero values (`0`, `""`, `[]`, `None`). `bool("False")` is `True`.
