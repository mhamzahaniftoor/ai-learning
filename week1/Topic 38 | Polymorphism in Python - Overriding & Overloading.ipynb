## Topic 38 | Polymorphism in Python: Overriding & Overloading

**Definition:**
Polymorphism means "many forms". The same method name or operator can
behave differently depending on the object or the arguments it gets.
It is the fourth pillar of OOP, after Encapsulation, Abstraction and
Inheritance.

**Real-life examples:**
- A "Pay" button on a shopping app: the same action works differently
  for Credit Card, JazzCash or Cash on Delivery.
- The `+` symbol: 5 + 3 gives 8 (numbers), "Hi" + "Ali" gives "HiAli"
  (strings), [1] + [2] gives [1, 2] (lists). Same symbol, different behaviour.
- A driver presses the same brake pedal in a car, a bus or a bike.
  The action is the same but the internal mechanism is different.

**Why use it:**
- The calling code stays simple: one method name for many object types.
- New types can be added without changing existing code.
- Programs become more flexible and easier to extend.

**Two types:**
1. Method Overriding (runtime polymorphism): the child class replaces
   the parent's method.
2. Method Overloading (compile-time polymorphism): the same method name
   with different parameters in the same class.

**1. Method Overriding (fully supported in Python)**

The child class gives its own version of a method that exists in the parent. Python decides which version to run based on the object's actual class.

Real-time example: payment system

```python
class Payment:
    def pay(self, amount):
        print(f"Paying Rs.{amount} using a generic method")

class CreditCard(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} using Credit Card")

class JazzCash(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} using JazzCash")

class CashOnDelivery(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} in cash at delivery")

# One loop, one method name, different behaviour for each object
methods = [CreditCard(), JazzCash(), CashOnDelivery()]

for method in methods:
    method.pay(5000)
```

The loop doesn't need to know which payment type it has. It just calls pay(). If you add EasyPaisa tomorrow, the loop doesn't change at all. This is the real power of polymorphism.

**2. Method Overloading (not supported the traditional way)**

In languages like Java or C++, you can write several methods with the same name but different parameters in one class. Python does not allow this. If you define the same method name twice, the second definition replaces the first:

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):          # this REPLACES the one above
        return a + b + c

calc = Calculator()
print(calc.add(2, 3, 4))   # 9 (works)
print(calc.add(2, 3))      # TypeError: missing 1 required argument: 'c'
```

Python's ways to get the same result:

Way 1: default arguments (simplest, most common)

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(2, 3))      # 5
print(calc.add(2, 3, 4))   # 9
```

**Way 2**: *args (any number of arguments)

```python
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print(calc.add(2, 3))          # 5
print(calc.add(2, 3, 4))       # 9
print(calc.add(1, 2, 3, 4, 5)) # 15
```

**Way 3**: check the argument type

```python
class Greeter:
    def greet(self, value=None):
        if value is None:
            print("Hello!")
        elif isinstance(value, str):
            print(f"Hello, {value}!")
        elif isinstance(value, int):
            print(f"Hello, you are {value} years old!")

g = Greeter()
g.greet()          # Hello!
g.greet("Ali")     # Hello, Ali!
g.greet(22)        # Hello, you are 22 years old!
```

**3. Operator Overloading (a form of overloading Python does support)**

Special methods with double underscores (dunder methods) let your own classes work with operators like +, == and len():

```python
class Wallet:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):                  # defines what + means
        return Wallet(self.amount + other.amount)

    def __str__(self):                         # defines what print() shows
        return f"Wallet(Rs.{self.amount})"

w1 = Wallet(500)
w2 = Wallet(300)
print(w1 + w2)    # Wallet(Rs.800)
```

Without __add__, w1 + w2 would give a TypeError. You taught Python what + means for a Wallet.

## Overriding vs Overloading

| | Overriding | Overloading |
|---|---|---|
| Where | Parent and child class | Same class |
| Method name | Same | Same |
| Parameters | Same | Different |
| Python support | Yes, built in | No, simulated with defaults or *args |
| Type | Runtime polymorphism | Compile-time polymorphism |
| Needs inheritance | Yes | No |

**Key points to remember:**
- Polymorphism = one name, many forms.
- Overriding: the child redefines the parent's method. Python picks
  the version based on the object's actual class.
- Overloading: Python keeps only the last definition of a name, so use
  default arguments or *args instead.
- Operator overloading uses dunder methods like __add__ and __str__.
- Use super().method() inside an override to extend the parent's
  behaviour instead of replacing it.

One more form, duck typing. Python polymorphism doesn't strictly need inheritance. If an object has the needed method, it works:

```python
class Dog:
    def speak(self):
        print("Woof")

class Robot:
    def speak(self):
        print("Beep boop")

for thing in [Dog(), Robot()]:
    thing.speak()     # works for both, with no shared parent
```

The idea is "if it speaks like a duck, it's a duck". Python only cares that speak() exists, not what class the object belongs to.
