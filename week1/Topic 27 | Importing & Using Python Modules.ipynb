## Modules in Python

**What is a module:**
A module is a large file containing Python code — variables and functions 
— that you can reuse.

**Importing a module:**
We use the `import` keyword to bring in a module, like the `math` module, 
and then use its functions.

```python
import math
print(math.sqrt(45))
```
Here, we need the dot (`math.sqrt`) because we imported the whole module, 
and `sqrt` belongs to it.

**Importing only specific functions — using `from`:**
```python
from math import sqrt
print(sqrt(45))
```
Here, we don't need the dot before `sqrt`, because we imported that 
function directly, not the whole module.

**Creating your own module:**
We can create our own custom file (saved with a `.py` extension) and 
use it like a module — importing its functions into another file.

````python
import my_module
````

````python
my_module.greeting()
````

````python
from my_module import greeting
````

````python
greeting()
````
