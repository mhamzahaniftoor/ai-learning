## Object-Oriented Programming (OOP)

**The problem OOP solves:**
As code grows, understanding and managing large amounts of code becomes 
difficult. We use the concept of encapsulation to deal with this.

**Encapsulation:**
Encapsulation means we bundle data and the functions that work on that 
data into a single unit, called an OBJECT.
- Data is encapsulated (stored) inside a single object.
- Functions related to that data are grouped together with it.
- This combination of grouped data + functions = Object-Oriented Programming.

**Object Attributes:**
An object has:
- Attributes (variables) — used to store data.
- Methods (functions) — that work on those variables within the object.

**Data Abstraction:**
We don't need to know HOW all of an object's internal data/logic works 
— just like a computer: it has complex circuitry inside, but we don't 
need to see that — we just interact with the outer shell (keyboard, 
screen, ports).

**Benefits of OOP:**
- **Promotes code readability** — code is organized into logical objects, making it easier to read and understand at a glance.
- **Enhances modularity** — each object is a self-contained unit, so code can be developed, tested, and reused independently.
- **Improves maintainability** — changes to one object's internals don't break unrelated parts of the program, making updates and bug fixes easier.
- **Makes programs closer to real-world thinking/modeling** — objects represent real-world entities (like a Car or Student), so the code structure matches how we naturally think about the problem.

**Example — Car (as an object):**
- Attributes: make, model, speed, color, etc.
- Methods: accelerate(), brake(), honk(), etc. — functions related to 
  the car, grouped inside the same object.

**Downside without OOP:**
Programs become difficult to debug and maintain when they don't follow 
OOP structuring — logic and data end up scattered instead of organized.

```python
class Car:
    def __init__(self, make, model, speed):
        self.make = make       # attribute
        self.model = model     # attribute
        self.speed = speed     # attribute

    def accelerate(self):      # method
        self.speed += 10
        print(f"{self.make} {self.model} speed: {self.speed}")

my_car = Car("Toyota", "Corolla", 0)
my_car.accelerate()   # Toyota Corolla speed: 10
```
