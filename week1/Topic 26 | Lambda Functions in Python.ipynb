## Lambda Functions

**What it is:**
Sometimes we need a function that has just a single line of code 
for a simple purpose. For that, we use a lambda function.

**Why use it:**
Basically used for small and trivial tasks, using a single line 
of code instead of writing a full function with `def`.

**Common use case:**
Often used together with built-in functions like `map()`, 
`filter()`, and `sorted()`.

```python
lambda arguments: expression
```

```python
# Regular function
def square(x):
    return x * x

# Same thing as a lambda
square = lambda x: x * x

print(square(5))   # 25
```

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x * x, numbers))
print(squared)   # [1, 4, 9, 16]

```

```python
numbers = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)   # [2, 4, 6]
```

```python
add = lambda a,b : a+b
```

```python
mul = lambda x: x*2
```

```python
mul(20)
```

```python
add(2,4)
```

```python
num = [1,2,4,5]
list(map(lambda num: num*num, num))
```

```python
num = [1,2,3,4,5,6,7,8,9]
list(filter(lambda x: x%2 != 0, num))
```
