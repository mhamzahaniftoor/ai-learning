## Topic 33 | Accessing Object Attributes & Methods

**Accessing attributes:**
Use dot notation: object.attribute
Example: student1.name → "Ali"
- No brackets, because an attribute is just stored data.
- Each object returns its own value (student1.name is "Ali",
  student2.name is "Sara").

**Calling instance methods:**
Use dot notation with brackets: object.method()
Example: student1.is_eligible()
- Brackets are required, because a method is a function that must be called.
- Python passes the object as `self` automatically, so you never write it
  in the call.

**Attribute vs method:**
- student1.name → attribute (data), no ()
- student1.is_eligible() → method (action), needs ()

**Common mistakes:**
- student1.is_eligible without () → does not run the method, it only
  refers to it.
- student1.name() → error, because name is data, not a function.

```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def is_eligible(self):
        if self.age > 18:
            print(f"{self.name} is Eligible for the course")
        else:
            print(f"{self.name} is Not Eligible")

student1 = Student("Ali", 20, "A")
student2 = Student("Sara", 16, "B")

# Accessing attributes (no brackets)
print(student1.name)    # Ali
print(student1.age)     # 20
print(student2.grade)   # B

# Calling instance methods (with brackets)
student1.is_eligible()  # Ali is Eligible for the course
student2.is_eligible()  # Sara is Not Eligible
```
