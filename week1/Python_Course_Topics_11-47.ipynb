# Topic 11 | Python Variables & Naming Conventions

## What is a variable?
A variable is a **container (a named box) that stores data**. In `age = 19`, we store `19` in the variable `age`.

It is called a *variable* because the value can **change**: `age = 20` today, `age = 21` after a year. So we don't only store data, we also update it.

## Dynamic typing
- We **don't declare a type** (no `int age`). We only write the name and the value.
- Python decides the type **from the value, at run time**. This is called a **dynamically typed** language.
- `age = 21` is an `int`, `gpa = 3.0` is a `float`, `name = "hamza"` is a `str`, `is_student = True` is a `bool`.
- The same variable can later hold a different type of data.
- Check a type with `type(variable)`.

## Naming rules (Python enforces these)
- Start with a **letter or underscore**, never a digit or special character.
- After the first character, use letters, digits and underscores only.
- **No spaces** (use `_` instead). No `-` or `$`.
- **Case-sensitive:** `age` and `Age` are different variables.
- Don't use **reserved words** (35 keywords, such as `class`, `if`, `for`, `True`).

## Naming style (good practice)
- Use **descriptive names** (`student_name`, not `x`), so the purpose is clear.
- Use **snake_case** for variables and functions: `student_name`.

````python
age = 21                    # int
gpa = 3.0                   # float
name = "hamza"              # str
is_student = True           # bool

print(type(age), type(gpa), type(name), type(is_student))

age = 22                    # change the value
print(age)

age = "twenty-two"          # change the value AND the type (allowed)
print(age, type(age))
````

````python
# Valid names
_x = 1
student_name = "Ali"
name2 = 1

# Case-sensitive: two different variables
Age = 5
age = 6
print(Age, age)             # 5 6

# Assign several at once, and swap
a, b = 5, 10
a, b = b, a
print(a, b)                 # 10 5

# Invalid names (each would give a SyntaxError):
# 2name = 1       starts with a digit
# my-name = 1     has a hyphen
# my name = 1     has a space
# class = 1       reserved word
````

## Valid vs invalid names

| Name | Valid? | Why |
|---|---|---|
| `student_name` | Yes | snake_case, recommended |
| `studentName` | Yes | allowed, but not Python style |
| `_count` | Yes | may start with an underscore |
| `name2` | Yes | digits are allowed after the first character |
| `2name` | **No** | starts with a digit |
| `my-name` | **No** | hyphen is a special character |
| `my name` | **No** | contains a space |
| `class` | **No** | reserved word |

## Key Points
- A variable is a named container whose value can **change**.
- No type declaration is needed. Python picks the type from the value at run time (**dynamic typing**).
- One variable can hold different types at different times, but keep one purpose per variable to avoid confusion.
- Names start with a letter or `_`, have no spaces or special characters, are case-sensitive, and can't be reserved words.
- Use descriptive **snake_case** names.

# Topic 12 | Python Data Types: Numbers, Strings & Booleans

A **data type** tells Python what kind of data a variable holds, and what operations are allowed on it.

## 1. Integer (`int`)
- Stores **whole numbers** (no decimal point): `19`, `-5`, `0`.
- Python integers have **no size limit**.

## 2. Float (`float`)
- Stores numbers **with a decimal point**: `3.0`, `3.5`, `-0.25`.
- `3` is an `int`, but `3.0` is a `float`.
- Division `/` always gives a float: `10 / 5` is `2.0`.

## 3. String (`str`)
- Stores **text**: letters, digits, spaces and special characters, written inside quotes (`"Ali"` or `'Ali'`).
- `"19"` in quotes is a string, not a number.
- Operations: `+` **concatenates** (joins), `*` **repeats**, `len()` gives the length, and `text[0]` gets one character.
- Strings are **immutable**: you can't change one character in place.
- `+` works only between two strings. `"Age: " + 19` gives a `TypeError`, so use `str(19)` or an f-string.

## 4. Boolean (`bool`)
- Stores only `True` or `False` (capital first letter).
- Often used as a **flag**: a switch that says yes or no, such as `is_valid = True`.
- Comes from comparisons: `5 > 3` is `True`.

````python
age = 19                 # int
gpa = 3.5                # float
is_valid = True          # bool

print(type(age), type(gpa), type(is_valid))

print(age + 1)           # 20
print(gpa * 2)           # 7.0
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)   # 3.5 3 1 8

print(10 / 5)            # 2.0 (division gives a float)
print(5 > 3, 5 == 3)     # True False

flag = False
if not flag:
    print("flag is off")
````

````python
name = "Ali"
father_name = "Hamza"

print(name + " " + father_name)     # Ali Hamza (concatenation)
print(name * 3)                     # AliAliAli (repetition)
print(len(name), name[0], name[-1], name.upper())   # 3 A i ALI

print("Age: " + str(19))            # convert the number to a string first
print(f"{name} is {age} years old") # f-string, the easiest way

print("19" + "1", 19 + 1)           # 191 20 (string vs number)

try:
    print("Age: " + 19)
except TypeError as e:
    print("TypeError:", e)
````

## Summary

| Type | Stores | Example | `type()` |
|---|---|---|---|
| Integer | Whole numbers | `19` | `int` |
| Float | Numbers with a decimal point | `3.5` | `float` |
| String | Text (letters, digits, spaces, symbols) | `"Ali"` | `str` |
| Boolean | `True` or `False` | `True` | `bool` |

## Key Points
- **Integer** = whole numbers. **Float** = decimal numbers. **String** = text in quotes. **Boolean** = `True` / `False`.
- `+` adds numbers but **joins** strings. You can't mix a string and a number with `+`, so convert with `str()` or use an f-string.
- A number in quotes (`"19"`) is a **string**.
- Floats can show tiny rounding errors: `0.1 + 0.2` gives `0.30000000000000004`. Use `round(x, 2)` when you need a clean result.
- `True` and `False` need a capital letter.
- Check any value's type with `type(value)`.

````python
print(round(0.1 + 0.2,))
````

````python
round(5.678, 1)
````

# Topic 13 | Python Data Structures - Lists, Tuples, Sets & Dictionaries

## What is a collection?
So far, one variable held **one value** (`age = 19`). A **collection** is a single variable that holds **many values**, for example all the fruits in a shop, or all the details of a student.

## Why use collections?
- **Less code:** one variable instead of `fruit1`, `fruit2`, `fruit3`, ...
- **Easy to process:** loop over all the items, search them, sort them.
- **Organized data:** related values stay together.
- **Flexible:** the size can grow or shrink while the program runs (for lists, sets and dictionaries).

## The 4 built-in collection types

| Type | Brackets | Ordered | Changeable (mutable) | Duplicates |
|---|---|---|---|---|
| **List** | `[ ]` | Yes | Yes | Allowed |
| **Tuple** | `( )` | Yes | **No** (immutable) | Allowed |
| **Set** | `{ }` | **No** | Yes (add/remove) | **Not allowed** |
| **Dictionary** | `{key: value}` | Yes (Python 3.7+) | Yes | Keys must be unique |

## Two words to know
- **Ordered:** items keep their position, so we can use an index (`fruits[0]`).
- **Mutable:** the collection can be changed after it is created. **Immutable** means it can't.

## 1. Lists
A list is an **ordered, changeable** collection written in square brackets. It can hold **multiple values**, including **duplicates** and mixed types.

    fruits = ["orange", "banana", "mango", "banana"]

## Index (position of an item)
- Counting starts from **0**, not 1.
- **Positive** index goes **left to right**: `0, 1, 2, 3 ...`
- **Negative** index goes **right to left**: `-1` is the last item, `-2` the second last.

| Item | orange | banana | mango | banana |
|---|---|---|---|---|
| Positive index | 0 | 1 | 2 | 3 |
| Negative index | -4 | -3 | -2 | -1 |

## Other basics
- **Slicing:** `fruits[1:3]` gives items at index 1 and 2 (the stop index is **not** included).
- **Change an item:** `fruits[0] = "apple"` (lists are mutable).
- **Length:** `len(fruits)`. **Check inside:** `"mango" in fruits`.
- Using an index that doesn't exist gives an `IndexError`.

````python
fruits = ["orange", "banana", "mango", "banana"]   # duplicates allowed

print(fruits)
print(type(fruits), len(fruits))

# Positive index: left to right from 0
print(fruits[0])      # orange
print(fruits[1])      # banana

# Negative index: right to left from -1
print(fruits[-1])     # banana (last item)
print(fruits[-2])     # mango

# Lists are mutable: change an item in place
fruits[0] = "apple"
print(fruits)

# Slicing [start:stop], stop is not included
print(fruits[1:3])    # ['banana', 'mango']
print(fruits[:2])     # first two
print(fruits[-2:])    # last two

# Membership and looping
print("mango" in fruits)
for fruit in fruits:
    print(fruit)

# A list can mix types
mixed = [1, "two", 3.0, True]
print(mixed)

# An index that doesn't exist
try:
    print(fruits[10])
except IndexError as e:
    print("IndexError:", e)
````

## List methods

| Method | What it does |
|---|---|
| `append(x)` | Adds `x` at the **end** |
| `insert(i, x)` | Adds `x` at index `i`; the items after it shift **right** |
| `remove(x)` | Removes the **first** item equal to `x` (not all of them). Error if not found. Items after it shift **left** |
| `pop()` | Removes **and returns** the last item |
| `pop(i)` | Removes **and returns** the item at index `i` |
| `extend(list2)` | Adds **all** items of another list at the end |
| `count(x)` | How many times `x` appears |
| `index(x)` | Index of the first `x` |
| `sort()` | Sorts the list **in place** |
| `reverse()` | Reverses the list **in place** |
| `clear()` | Removes everything |
| `copy()` | Makes a real copy |

## Notes
- `sort()` changes the original list. `sorted(list)` returns a **new** sorted list and leaves the original alone.
- `append` adds **one** item. To add several, use `extend`.
- Text is case-sensitive: `"orange"` and `"Orange"` are different items.

````python
fruits = ["orange", "banana", "mango", "banana"]

fruits.append("kiwi")              # add at the end
print(fruits)

fruits.insert(1, "grapes")         # add at index 1
print(fruits)

fruits.remove("banana")            # removes only the FIRST "banana"
print(fruits)

last = fruits.pop()                # removes AND returns the last item
print(last, fruits)

second = fruits.pop(1)             # pop by index
print(second, fruits)

fruits.extend(["lemon", "peach"])  # add many items
print(fruits)

print(fruits.count("orange"), fruits.index("mango"))

fruits.sort()                      # sort in place
print(fruits)
fruits.reverse()
print(fruits)
print(sorted(fruits))              # new sorted list, original unchanged

try:
    fruits.remove("pineapple")     # not in the list
except ValueError as e:
    print("ValueError:", e)

fruits.clear()
print(fruits)
````

````python
a = [1, 2, 3]
b = a                # NOT a copy: both names point to the SAME list
b.append(4)
print(a)             # [1, 2, 3, 4]  a changed too!

c = a.copy()         # a real, separate copy
c.append(5)
print(a, c)          # [1, 2, 3, 4] [1, 2, 3, 4, 5]
````

## 2. Tuples
A tuple is an **ordered but unchangeable (immutable)** collection written in **round brackets**. Once created, you can't add, remove or change items. Duplicates are allowed.

    colors = ("pink", "green", "brown", "pink")

## Why use a tuple instead of a list?
- **Protection:** the data can't be changed by accident (days of the week, coordinates, fixed settings).
- **Slightly faster and lighter** than a list.
- Can be used as a **dictionary key** (lists can't).
- Handy for returning several values from a function.

## Rules and notes
- Index and slicing work exactly like lists: `colors[1]`, `colors[-1]`.
- Trying to change an item gives a `TypeError`.
- **Only two methods:** `count(x)` and `index(x)`.
- A tuple with **one** item needs a comma: `("pink",)`. Without the comma, `("pink")` is just a string.
- **Unpacking:** `r, g, b = (255, 128, 0)` puts each value in its own variable.
- To "change" a tuple, convert it: `list(t)` -> edit -> `tuple(...)`.
- **Careful:** curly brackets `{ }` make a **set**, not a tuple.

````python
colors = ("pink", "green", "brown", "pink")      # round brackets
print(colors, type(colors))

print(colors[1])         # green
print(colors[-1])        # pink

# Immutable: changing an item fails
try:
    colors[1] = "black"
except TypeError as e:
    print("TypeError:", e)

# The only two methods
print(colors.count("pink"))      # 2
print(colors.index("green"))     # 1

# One-item tuple needs a comma
one = ("pink",)
not_tuple = ("pink")
print(type(one), type(not_tuple))

# Unpacking
r, g, b = (255, 128, 0)
print(r, g, b)

# "Changing" a tuple: convert to list, edit, convert back
temp = list(colors)
temp[1] = "black"
colors = tuple(temp)
print(colors)

# Tuples can be dictionary keys
location = {(31.5, 74.3): "Lahore"}
print(location[(31.5, 74.3)])

# Curly brackets make a SET, not a tuple
wrong = {"pink", "green", "brown", "pink"}
print(type(wrong), len(wrong))
try:
    wrong[1]
except TypeError as e:
    print("TypeError:", e)
````

## 3. Sets
A set is an **unordered** collection of **unique** items, written in curly brackets. **Duplicates are removed automatically.**

    numbers = {1, 2, 3, 4, 4}      # becomes {1, 2, 3, 4}

## Key properties
- **Unordered:** no positions, so **no indexing** (`numbers[0]` gives a `TypeError`) and no guaranteed print order.
- **Unique:** a duplicate is silently ignored.
- **Fast to search:** `x in my_set` is much faster than in a list, because sets use hashing to find items directly.
- **Mutable:** you can add and remove items, but the items themselves must be immutable (numbers, strings, tuples; **not** lists).
- **Empty set:** use `set()`. Writing `{}` creates an empty **dictionary**.
- Don't name a variable `set`, it hides the built-in.

## Methods

| Method | What it does |
|---|---|
| `add(x)` | Adds `x` (nothing happens if it's already there) |
| `remove(x)` | Removes `x`. **KeyError** if it isn't there |
| `discard(x)` | Removes `x`. **No error** if it isn't there |
| `pop()` | Removes and returns a random item |
| `clear()` | Removes everything |

## Best use: remove duplicates
`set(my_list)` gives the unique items of a list in one step.

````python
numbers = {1, 2, 3, 4, 4}          # the duplicate 4 is dropped
print(numbers, type(numbers))

numbers.add(5)
numbers.remove(1)
print(numbers)                      # {2, 3, 4, 5}

numbers.add(3)                      # already there, nothing happens
print(numbers)

try:
    numbers.remove(99)              # not there
except KeyError as e:
    print("KeyError:", e)
numbers.discard(99)                 # no error

try:
    print(numbers[0])               # sets have no index
except TypeError as e:
    print("TypeError:", e)

print(3 in numbers)                 # fast membership test

empty_set = set()
empty_dict = {}
print(type(empty_set), type(empty_dict))

# Remove duplicates from a list
names = ["Ali", "Sara", "Ali", "Ahmed", "Sara"]
unique_names = set(names)
print(sorted(unique_names), len(unique_names))
````

## Set operations
Take two sets `a` and `b`:

| Operation | Method | Operator | Meaning |
|---|---|---|---|
| Union | `a.union(b)` | `a \| b` | All items from both |
| Intersection | `a.intersection(b)` | `a & b` | Only items in **both** |
| Difference | `a.difference(b)` | `a - b` | Items in `a` but **not** in `b` |
| Symmetric difference | `a.symmetric_difference(b)` | `a ^ b` | Items in **either**, but not both |
| Subset | `a.issubset(b)` | | Is every item of `a` also in `b`? |
| Disjoint | `a.isdisjoint(b)` | | Do they share **nothing**? |

Note: `a.difference(b)` is **not** the same as `b.difference(a)`. Order matters.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a.union(b))                  # {1, 2, 3, 4, 5, 6}
print(a.intersection(b))           # {3, 4}
print(a.difference(b))             # {1, 2}
print(b.difference(a))             # {5, 6}
print(a.symmetric_difference(b))   # {1, 2, 5, 6}

print({1, 2}.issubset(a))          # True
print(a.isdisjoint({7, 8}))        # True

# Same results with operators
print(a | b, a & b, a - b, a ^ b)
````

## 4. Dictionaries
A dictionary stores data as **key: value pairs**, like a real dictionary (word -> meaning) or a student record (`"name"` -> `"hamza"`).

    student = {"name": "hamza", "grade": "A"}

## Key properties
- Each value is found by its **key**, not by a position. This makes lookup **very fast**.
- **Keys must be unique.** If a key repeats, the last value wins.
- **Keys must be immutable** (strings, numbers, tuples). Values can be anything, including lists or other dictionaries.
- **Mutable:** add, change and remove pairs at any time.
- Keeps **insertion order** (Python 3.7+).

## Common operations

| Operation | Code | Notes |
|---|---|---|
| Read a value | `student["name"]` | `KeyError` if the key is missing |
| Read safely | `student.get("grade")` | Gives `None` if missing, or your own default: `.get("age", "unknown")` |
| Add / change | `student["age"] = 21` | Adds if new, replaces if it exists |
| Remove | `student.pop("grade")` | Removes the key **and** value, returns the value |
| Remove | `del student["age"]` | Removes the pair |
| Check a key | `"name" in student` | Checks **keys**, not values |
| Merge | `student.update({...})` | Adds or replaces several pairs |
| Size | `len(student)` | Number of pairs |
| All keys | `student.keys()` | |
| All values | `student.values()` | |
| All pairs | `student.items()` | Each pair is a `(key, value)` tuple |

````python
student = {"name": "hamza", "grade": "A"}
print(student, type(student))

print(student["name"])                   # hamza
print(student.get("grade"))              # A
print(student.get("age"))                # None (no error)
print(student.get("age", "unknown"))     # custom default

try:
    print(student["age"])                # missing key
except KeyError as e:
    print("KeyError:", e)

student["name"] = "M Hamza"              # change an existing value
student["age"] = 21                      # add a new pair
print(student)

removed = student.pop("grade")           # removes key and value
print(removed, student)

del student["age"]
print(student)

print("name" in student)                 # True (checks keys)
print(len(student))                      # 1
````

````python
student = {"name": "M Hamza", "grade": "A", "age": 21}

print(student.keys())
print(student.values())
print(student.items())

for key in student:                      # loops over the keys
    print(key)

for value in student.values():           # loops over the values
    print(value)

for key, value in student.items():       # loops over both (most common)
    print(key, ":", value)

# Looping without .items() and unpacking fails
try:
    for key, value in student:
        print(key, value)
except ValueError as e:
    print("ValueError:", e)

# update(): change one pair and add another
student.update({"grade": "B", "city": "Lahore"})
print(student)

# Nested dictionary: a dictionary inside a dictionary
students = {
    "s1": {"name": "Ali", "grade": "A"},
    "s2": {"name": "Sara", "grade": "B"},
}
print(students["s2"]["name"])            # Sara

# Duplicate keys: the last value wins
d = {"a": 1, "a": 2}
print(d)
````

````python
# A list of dictionaries (the shape of most real data, like JSON from an API)
students = [
    {"name": "Ali", "grade": "A", "age": 20},
    {"name": "Sara", "grade": "B", "age": 16},
    {"name": "Ahmed", "grade": "A", "age": 22},
]

for s in students:
    print(s["name"], s["grade"])

# Build a list of the top students
top_students = []
for s in students:
    if s["grade"] == "A":
        top_students.append(s["name"])
print(top_students)

# Build a set of the unique grades
grades = set()
for s in students:
    grades.add(s["grade"])
print(sorted(grades))

# Converting between collections
numbers = [3, 1, 2, 3, 1]
print(tuple(numbers), set(numbers), sorted(set(numbers)))

pairs = [("a", 1), ("b", 2)]
print(dict(pairs))
print(list("abc"), list({"x": 1, "y": 2}))
````

## Comparison of all four

| | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Brackets | `[ ]` | `( )` | `{ }` | `{k: v}` |
| Ordered | Yes | Yes | **No** | Yes (3.7+) |
| Mutable | Yes | **No** | Yes | Yes |
| Duplicates | Allowed | Allowed | **Not allowed** | Keys unique, values can repeat |
| Access by | Index | Index | No direct access (use `in`, loop) | **Key** |
| Fast lookup | No (searches one by one) | No | **Yes** | **Yes (by key)** |
| Typical use | Shopping list, records | Fixed data, coordinates | Unique items, removing duplicates | Records, settings, lookups |

## Which one should I use?
- Items in order that will **change** -> **List**
- Items in order that must **not change** -> **Tuple**
- Only **unique** items, or you need set math (union, intersection) -> **Set**
- Each value has a **name or label** -> **Dictionary**

## Key Points
- A collection stores **many values in one variable**.
- **List** = ordered + mutable + duplicates. **Tuple** = ordered + immutable. **Set** = unordered + unique. **Dictionary** = key-value pairs.
- Indexes start at **0**. Negative indexes count from the **right**, starting at `-1`.
- `[ ]` list, `( )` tuple, `{ }` set or dictionary. An empty set is `set()`, not `{}`.
- Tuples have only `count` and `index`. Sets have no index. Dictionaries use keys.
- `b = a` does **not** copy a list. Use `a.copy()`.
- Use `.get()` on a dictionary when the key might be missing, and `.items()` to loop over keys and values together.

# Topic 14 | Type Conversion & Casting in Python

Converting a value from one data type to another, for example integer to string, string to integer, integer to float, or float to integer.

## Two kinds
- **Implicit conversion (type conversion):** Python does it **automatically** when it is safe. `5 + 2.5` gives `7.5`, because the int `5` becomes `5.0` first.
- **Explicit conversion (type casting):** **we** do it using a conversion function.

## Casting functions

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int(x)` | Integer | `int("25")`, `int(3.9)` | `25`, `3` (decimal part is **cut off**, not rounded) |
| `float(x)` | Float | `float("3.5")`, `float(5)` | `3.5`, `5.0` |
| `str(x)` | String | `str(100)`, `str(3.0)` | `"100"`, `"3.0"` |
| `bool(x)` | Boolean | `bool(0)`, `bool("hi")` | `False`, `True` |

## Why do we need it?
- `input()` always returns a **string**, so numbers must be converted before using them in calculations.
- `"Age: " + 34` fails. Both sides of `+` must be strings, so convert with `str(34)`.

# Topic 15 | Python Operators: Arithmetic, Comparison, Logical & Assignment

An **operator** is a symbol that performs an operation on values. The values it works on are called **operands**. In `2 + 3`, the operator is `+` and the operands are `2` and `3`.

## By number of operands
- **Unary** operator: **one** operand. Examples: `-5` (negative), `+5`, `not True`.
- **Binary** operator: **two** operands. Examples: `2 + 3`, `5 > 2`, `a and b`.

## By purpose: 4 groups

### 1. Arithmetic (maths)
| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `2 + 3` | `5` |
| `-` | Subtraction | `2 - 3` | `-1` |
| `*` | Multiplication | `2 * 6` | `12` |
| `/` | Division (always a float) | `2 / 9` | `0.2222...` |
| `//` | Floor division (rounds **down**) | `2 // 9` | `0` |
| `%` | Modulus (remainder) | `101 % 10` | `1` |
| `**` | Power | `2 ** 4` | `16` |

`%` is handy for even/odd checks (`n % 2 == 0`), and `//` with `%` splits a number into quotient and remainder. Floor division rounds toward **negative** infinity: `-7 // 2` is `-4`, not `-3`.

### 2. Comparison (the result is always `True` or `False`)
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `2 == 4` | `False` |
| `!=` | Not equal to | `2 != 2` | `False` |
| `>` | Greater than | `2 > 6` | `False` |
| `<` | Less than | `2 < 1` | `False` |
| `>=` | Greater than or equal | `2 >= 2` | `True` |
| `<=` | Less than or equal | `2 <= 1` | `False` |

Python allows **chained** comparisons: `1 < age < 100`.

### 3. Logical (combine True/False values)
| Operator | Result is `True` when | Type |
|---|---|---|
| `and` | **both** sides are `True` | binary |
| `or` | **at least one** side is `True` | binary |
| `not` | reverses the value (`not True` is `False`) | unary |

`and` / `or` use **short-circuit** evaluation: Python stops as soon as the answer is known. They return one of the operands, not always `True`/`False`: `0 or "default"` gives `"default"`.

### 4. Assignment (store or update a value)
| Operator | Same as |
|---|---|
| `=` | store a value: `x = 10` |
| `+=` | `x = x + 5` |
| `-=` | `x = x - 3` |
| `*=` | `x = x * 2` |
| `/=` | `x = x / 4` |
| `//=` | `x = x // 5` |
| `%=` | `x = x % 2` |
| `**=` | `x = x ** 3` |

````python
# Arithmetic (binary: two operands)
print(2 + 3)             # 5
print(2 - 3)             # -1
print(2 * 6)             # 12
print(2 / 9)             # 0.2222222222222222 (always a float)
print(2 // 9)            # 0 (floor division)
print(101 % 10)          # 1 (remainder)
print(2 ** 4)            # 16 (power)
print(-7 // 2, -7 % 2)   # -4 1 (floor division rounds DOWN)

# Unary (one operand)
x = 5
print(-x, +x, not True)  # -5 5 False

# Comparison (result is always True or False)
print(2 == 4)            # False
print(2 != 2)            # False
print(2 > 6)             # False
print(2 < 1)             # False
print(2 >= 2, 2 <= 1)    # True False

age = 25
print(1 < age < 100)     # True (chained comparison)
print("apple" == "apple", "a" < "b", 2 == 2.0)   # True True True
````

````python
age = 25
has_id = True

# Logical
print(age >= 18 and has_id)      # True  (both True)
print(age < 18 or has_id)        # True  (one is True)
print(not has_id)                # False (reversed)
print(0 or "default", 5 and 6)   # default 6 (they return an operand)

# Assignment operators
x = 10
x += 5; print(x)     # 15
x -= 3; print(x)     # 12
x *= 2; print(x)     # 24
x /= 4; print(x)     # 6.0
y = 17
y //= 5; print(y)    # 3
y %= 2; print(y)     # 1
y = 3
y **= 3; print(y)    # 27

# Precedence: which operator runs first
print(2 + 3 * 4)     # 14 (multiplication first)
print((2 + 3) * 4)   # 20 (brackets first)
print(2 ** 3 ** 2)   # 512 (power goes right to left: 2 ** 9)
print(-2 ** 2)       # -4 (power before the minus)

# A common trap with floats
print(0.1 + 0.2 == 0.3)               # False
print(round(0.1 + 0.2, 2) == 0.3)     # True
````

## Operator precedence (highest to lowest)

| Order | Operators |
|---|---|
| 1 | `( )` brackets |
| 2 | `**` |
| 3 | unary `+x`, `-x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | comparisons: `==`, `!=`, `>`, `<`, `>=`, `<=` |
| 7 | `not` |
| 8 | `and` |
| 9 | `or` |

When unsure, **use brackets**. They make the order clear to the reader.

## Key Points
- **Unary** = one operand (`-5`, `not x`). **Binary** = two operands (`2 + 3`).
- `/` always returns a float. `//` rounds down. `%` gives the remainder. `**` is power.
- Comparison operators always return `True` or `False`. Don't confuse `=` (assign) with `==` (compare).
- Logical operators are the words `and`, `or`, `not`.
- `x += 1` is shorthand for `x = x + 1`. Python has no `x++`.
- Floats can be slightly inexact (`0.1 + 0.2 != 0.3`), so use `round()` when comparing them.

# Topic 28 | Understanding the Need For OOP

## Object-Oriented Programming (OOP)

**The problem OOP solves:**
As code grows, understanding and managing large amounts of code becomes 
difficult. We use the concept of encapsulation to deal with this.

**Encapsulation:**
Encapsulation means we bundle data and the functions that work on that 
data into a single unit, called an OBJECT.
- Data is encapsulated (stored) inside a single object.
- Functions related to that data are grouped together with it.
- This combination of grouped data + functions = Object-Oriented Programming.

**Object Attributes:**
An object has:
- Attributes (variables) — used to store data.
- Methods (functions) — that work on those variables within the object.

**Data Abstraction:**
We don't need to know HOW all of an object's internal data/logic works 
— just like a computer: it has complex circuitry inside, but we don't 
need to see that — we just interact with the outer shell (keyboard, 
screen, ports).

**Benefits of OOP:**
- **Promotes code readability** — code is organized into logical objects, making it easier to read and understand at a glance.
- **Enhances modularity** — each object is a self-contained unit, so code can be developed, tested, and reused independently.
- **Improves maintainability** — changes to one object's internals don't break unrelated parts of the program, making updates and bug fixes easier.
- **Makes programs closer to real-world thinking/modeling** — objects represent real-world entities (like a Car or Student), so the code structure matches how we naturally think about the problem.

**Example — Car (as an object):**
- Attributes: make, model, speed, color, etc.
- Methods: accelerate(), brake(), honk(), etc. — functions related to 
  the car, grouped inside the same object.

**Downside without OOP:**
Programs become difficult to debug and maintain when they don't follow 
OOP structuring — logic and data end up scattered instead of organized.

````python
class Car:
    def __init__(self, make, model, speed):
        self.make = make       # attribute
        self.model = model     # attribute
        self.speed = speed     # attribute

    def accelerate(self):      # method
        self.speed += 10
        print(f"{self.make} {self.model} speed: {self.speed}")

my_car = Car("Toyota", "Corolla", 0)
my_car.accelerate()   # Toyota Corolla speed: 10
````

## The `__init__` Method (Constructor)

**Definition:**
`__init__` is a special method in a class called a CONSTRUCTOR. 
It is called AUTOMATICALLY by Python the moment an object is 
created — you never call it yourself directly.

**Purpose:**
Its job is to set up (initialize) the object's initial attributes 
— i.e., assign starting values to the object's data as soon as

**How it's different from a regular method:**
- `__init__` → runs automatically, immediately when the object is created.
- A regular method (like `start()`) → only runs when you explicitly

**Example:**
```python
class Car:
    def __init__(self, color):
        self.color = color
        print(f"A new {self.color} car object was created!")  # runs automatically

    def start(self):                                            # regular method
        print(f"{self.color} car started")

# __init__ runs automatically here, the moment the object is made
my_car = Car("red")
# Output: A new red car object was created!

# start() does NOT run automatically — only runs when explicitly called
my_car.start()
# Output: red car started
```

**Key takeaway:**
You will never see `my_car.__init__("red")` written in code — Python 
calls it for you behind the scenes as soon as you write `Car("red")`. 
Regular methods like `start()` always need to be called explicitly 
using dot notation.

````python
class Car:
    def __init__(self, color):
        self.color = color

    def start(self):
        print(f"{self.color} car started")

# First object
my_car = Car("red")
my_car.start()        # red car started

# Second object
mybrother_car = Car("red")
mybrother_car.start() # red car started
````

# Topic 32 | Instance Variables & Instance Methods

## Instances and Instance Methods

**Instance:**
An instance is an object created from a class. If Car is the class,
then each car we make from it (my_car, mybrother_car) is an instance.

**Each instance has its own data:**
Every object keeps its own attributes (e.g. make, model, size, color).
Changing one object's data does not affect another object.

**Instance methods:**
- An instance method is a function inside a class that takes `self`
  as its first parameter.
- `self` lets the method work with that specific object's own data.
- The method is defined once in the class and shared by all objects,
  but it uses each object's own attribute values, so results differ
  from object to object.
- If a function inside a class does not take `self`, it is not an
  instance method and cannot access the object's attributes.

````python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def is_eligible(self):
        if self.age > 18:
            print(f"{self.name} is Eligible for the course")
        else:
            print(f"{self.name} is Not Eligible")

student1 = Student("Ali", 20, "A")
student2 = Student("Sara", 16, "B")

student1.is_eligible()   # Ali is Eligible for the course
student2.is_eligible()   # Sara is Not Eligible
````

# Topic 34 | Encapsulation in Python - Public & Private Attributes

## Encapsulation and Private Attributes

**Encapsulation:**
Keeping an object's data (attributes) and functions (methods) together
inside the object, and exposing only what is needed. Like a mouse, we use
the left and right buttons (the functionality) without seeing the circuit
(the core logic).

**Private attributes:**
- A private attribute is meant to be used only inside the class, not
  from outside it.
- We make an attribute private by putting two underscores before its
  name: `__grade`.
- Outside code cannot read it directly (`student1.__grade` gives an error).
- We access it through a method instead. A method that reads the value
  is called a GETTER.

**Why use it:**
- Protects data from accidental or invalid changes.
- Lets the class control how its data is read and modified.
- Internal logic can change without breaking the code that uses the class.

````python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade        # private attribute

    def get_grade(self):            # getter
        return self.__grade

student1 = Student("Ali", "A")

print(student1.name)         # Ali (public, direct access works)
print(student1.get_grade())  # A (private, accessed through the getter)
print(student1.__grade)      # AttributeError: no attribute '__grade'
````

Setter: the natural partner of a getter. It lets you change a private value with validation, and this is where encapsulation pays off:

````python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade

    def get_grade(self):
        return self.__grade

    def set_grade(self, new_grade):
        if new_grade in ["A", "B", "C", "D", "F"]:
            self.__grade = new_grade
        else:
            print("Invalid grade")

student1 = Student("Ali", "A")
student1.set_grade("B")
print(student1.get_grade())   # B
student1.set_grade("Z")       # Invalid grade
````

# Topic 35 | Using Getters & Setters in Python

## Getter and Setter

**Getter:**
A method used to GET (read) the value of a private attribute from
outside the class.
Example: get_grade()

**Setter:**
A method used to SET (change) the value of a private attribute from
outside the class.
Example: set_grade()

**Why we need them:**
- A private attribute (like __grade) cannot be accessed directly
  from outside the class.
- Getter and setter are the controlled doors to that data.
- The setter can validate the new value before accepting it, so
  invalid data never gets in. Direct access cannot do this.

**Naming convention:**
get_attributeName() and set_attributeName()

````python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade              # private attribute

    # Getter: reads the private attribute
    def get_grade(self):
        return self.__grade

    # Setter: changes the private attribute, with validation
    def set_grade(self, new_grade):
        if new_grade in ["A", "B", "C", "D", "F"]:
            self.__grade = new_grade
        else:
            print("Invalid grade")

student1 = Student("Ali", "A")

print(student1.get_grade())   # A      (using the getter)

student1.set_grade("B")       # using the setter
print(student1.get_grade())   # B

student1.set_grade("Z")       # Invalid grade (setter rejected it)
print(student1.get_grade())   # B      (value unchanged)
````

# Topic 36 | Inheritance in Python - Parent & Child Classes

## Inheritance

**Definition:**
Inheritance lets one class (the child) automatically get the attributes
and methods of another class (the parent). It is an advanced form of
code reusability.

**Terms:**
- Parent class (also called base class or superclass): the class being inherited from, e.g. Animal
- Child class (also called derived class or subclass): the class that inherits, e.g. Dog, Cat

**How it works:**
- Write the shared behaviour once in the parent (e.g. speak() in Animal).
- Each child class gets it for free, so you don't rewrite it.
- Each child can also add its own unique behaviour (Dog has bark(), Cat has meow()).

**Syntax:**
class Child(Parent):
The parent's name goes in brackets after the child's name.

````python
class Animal:                       # parent class
    def speak(self):
        print("Speak")

class Dog(Animal):                  # child class
    def bark(self):
        print("Woof!")

class Cat(Animal):                  # another child class
    def meow(self):
        print("Meow!")

dog1 = Dog()
dog1.bark()     # Woof!   (Dog's own method)
dog1.speak()    # Speak   (inherited from Animal)

cat1 = Cat()
cat1.meow()     # Meow!   (Cat's own method)
cat1.speak()    # Speak   (inherited from Animal)

# dog1.meow()   # AttributeError: Dog doesn't have meow
````

Python looks in the child class first, and only goes to the parent if it doesn't find the method.

Inheritance with __init__ and super(): if the child has its own __init__, it should call the parent's __init__ using super(), otherwise the parent's attributes never get set:

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)      # runs Animal's __init__ for name
        self.breed = breed          # Dog's own attribute

    def bark(self):
        print(f"{self.name} says Woof!")

dog1 = Dog("Tommy", "Labrador")
dog1.speak()    # Tommy makes a sound
dog1.bark()     # Tommy says Woof!
print(dog1.breed)   # Labrador
````

# Topic 37 | Method Overriding

## Method Overriding

**Definition:**
Method overriding is when a child class defines a method with the
same name as a method in its parent class, so the child's version
replaces (overrides) the parent's version for the child's objects.

**Why use it:**
- The parent gives a general behaviour (e.g. every Animal can speak).
- Each child can change it to its own specific behaviour
  (a Dog barks, a Cat meows) without touching the parent.

**Rules:**
- The method name must be exactly the same in parent and child.
- Python looks in the child class first. If it finds the method there,
  it uses it and never goes to the parent.
- If the child does not define the method, the parent's version is used.

````python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):                # overrides Animal.speak
        print("Woof woof")

class Cat(Animal):
    def speak(self):                # overrides Animal.speak
        print("Meow")

class Cow(Animal):
    pass                            # no override, uses the parent's version

animal1 = Animal()
dog1 = Dog()
cat1 = Cat()
cow1 = Cow()

animal1.speak()   # Animal makes a sound
dog1.speak()      # Woof woof
cat1.speak()      # Meow
cow1.speak()      # Animal makes a sound (inherited, not overridden)
````

````python
class Dog(Animal):
    def speak(self):
        super().speak()             # run the parent's version first
        print("Woof woof")          # then add Dog's own behaviour

Dog().speak()
# Animal makes a sound
# Woof woof
````

# Topic 38 | Polymorphism in Python: Overriding & Overloading

**Definition:**
Polymorphism means "many forms". The same method name or operator can
behave differently depending on the object or the arguments it gets.
It is the fourth pillar of OOP, after Encapsulation, Abstraction and
Inheritance.

**Real-life examples:**
- A "Pay" button on a shopping app: the same action works differently
  for Credit Card, JazzCash or Cash on Delivery.
- The `+` symbol: 5 + 3 gives 8 (numbers), "Hi" + "Ali" gives "HiAli"
  (strings), [1] + [2] gives [1, 2] (lists). Same symbol, different behaviour.
- A driver presses the same brake pedal in a car, a bus or a bike.
  The action is the same but the internal mechanism is different.

**Why use it:**
- The calling code stays simple: one method name for many object types.
- New types can be added without changing existing code.
- Programs become more flexible and easier to extend.

**Two types:**
1. Method Overriding (runtime polymorphism): the child class replaces
   the parent's method.
2. Method Overloading (compile-time polymorphism): the same method name
   with different parameters in the same class.

**1. Method Overriding (fully supported in Python)**

The child class gives its own version of a method that exists in the parent. Python decides which version to run based on the object's actual class.

Real-time example: payment system

````python
class Payment:
    def pay(self, amount):
        print(f"Paying Rs.{amount} using a generic method")

class CreditCard(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} using Credit Card")

class JazzCash(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} using JazzCash")

class CashOnDelivery(Payment):
    def pay(self, amount):                  # overrides Payment.pay
        print(f"Paying Rs.{amount} in cash at delivery")

# One loop, one method name, different behaviour for each object
methods = [CreditCard(), JazzCash(), CashOnDelivery()]

for method in methods:
    method.pay(5000)
````

The loop doesn't need to know which payment type it has. It just calls pay(). If you add EasyPaisa tomorrow, the loop doesn't change at all. This is the real power of polymorphism.

**2. Method Overloading (not supported the traditional way)**

In languages like Java or C++, you can write several methods with the same name but different parameters in one class. Python does not allow this. If you define the same method name twice, the second definition replaces the first:

````python
class Calculator:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):          # this REPLACES the one above
        return a + b + c

calc = Calculator()
print(calc.add(2, 3, 4))   # 9 (works)
print(calc.add(2, 3))      # TypeError: missing 1 required argument: 'c'
````

Python's ways to get the same result:

Way 1: default arguments (simplest, most common)

````python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(2, 3))      # 5
print(calc.add(2, 3, 4))   # 9
````

**Way 2**: *args (any number of arguments)

````python
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print(calc.add(2, 3))          # 5
print(calc.add(2, 3, 4))       # 9
print(calc.add(1, 2, 3, 4, 5)) # 15
````

**Way 3**: check the argument type

````python
class Greeter:
    def greet(self, value=None):
        if value is None:
            print("Hello!")
        elif isinstance(value, str):
            print(f"Hello, {value}!")
        elif isinstance(value, int):
            print(f"Hello, you are {value} years old!")

g = Greeter()
g.greet()          # Hello!
g.greet("Ali")     # Hello, Ali!
g.greet(22)        # Hello, you are 22 years old!
````

**3. Operator Overloading (a form of overloading Python does support)**

Special methods with double underscores (dunder methods) let your own classes work with operators like +, == and len():

````python
class Wallet:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):                  # defines what + means
        return Wallet(self.amount + other.amount)

    def __str__(self):                         # defines what print() shows
        return f"Wallet(Rs.{self.amount})"

w1 = Wallet(500)
w2 = Wallet(300)
print(w1 + w2)    # Wallet(Rs.800)
````

Without __add__, w1 + w2 would give a TypeError. You taught Python what + means for a Wallet.

## Overriding vs Overloading

| | Overriding | Overloading |
|---|---|---|
| Where | Parent and child class | Same class |
| Method name | Same | Same |
| Parameters | Same | Different |
| Python support | Yes, built in | No, simulated with defaults or *args |
| Type | Runtime polymorphism | Compile-time polymorphism |
| Needs inheritance | Yes | No |

**Key points to remember:**
- Polymorphism = one name, many forms.
- Overriding: the child redefines the parent's method. Python picks
  the version based on the object's actual class.
- Overloading: Python keeps only the last definition of a name, so use
  default arguments or *args instead.
- Operator overloading uses dunder methods like __add__ and __str__.
- Use super().method() inside an override to extend the parent's
  behaviour instead of replacing it.

One more form, duck typing. Python polymorphism doesn't strictly need inheritance. If an object has the needed method, it works:

````python
class Dog:
    def speak(self):
        print("Woof")

class Robot:
    def speak(self):
        print("Beep boop")

for thing in [Dog(), Robot()]:
    thing.speak()     # works for both, with no shared parent
````

The idea is "if it speaks like a duck, it's a duck". Python only cares that speak() exists, not what class the object belongs to.

You have now covered all four OOP pillars: Encapsulation, Abstraction, Inheritance and Polymorphism.

# Topic 39 | Function vs Method vs Special Method

## Function
A block of code defined on its own, usually at the top level of the program (global context). It belongs to no class and is called directly by name: `greet()`.

## Method
A method is also a function, but it is **defined inside a class**. It belongs to the class and works on the object's data.
- The first parameter is `self`.
- It is called through an object: `obj.method()`.

## Special Method (Magic / Dunder Method)
A built-in method with a fixed name that starts and ends with double underscores, such as `__init__`.
- Python calls it **automatically** when something happens. You don't call it yourself.
- You define it in your class to control how your objects behave with built-in features (`print()`, `len()`, `+`, `==`).

## Function vs Method

| | Function | Method |
|---|---|---|
| Defined | Outside any class | Inside a class |
| Belongs to | Nobody (standalone) | A class and its objects |
| First parameter | None required | `self` |
| Called with | `name()` | `object.name()` |
| Can use object data | No | Yes, through `self` |

````python
# Function: standalone, global context
def greet(name):
    print(f"Hello, {name}")

greet("Ali")
````

````python
# Method: defined inside a class
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):            # method
        print(f"Hello, {self.name}")

s1 = Student("Sara")
s1.greet()                      # called through the object
````

## Common Special Methods

| Method | Called automatically when | Purpose |
|---|---|---|
| `__init__` | an object is created | Constructor, sets the initial attributes |
| `__str__` | `print(obj)` or `str(obj)` | Readable text for the object |
| `__repr__` | the object is shown in the console, or `repr(obj)` | Developer-friendly text for the object |
| `__len__` | `len(obj)` | Defines the object's length |
| `__add__` | `obj1 + obj2` | Defines what `+` means |
| `__eq__` | `obj1 == obj2` | Defines when two objects are equal |

````python
class Book:
    def __init__(self, title, pages):      # runs when Book(...) is created
        self.title = title
        self.pages = pages

    def __str__(self):                     # runs on print(book)
        return f"Book: {self.title}"

    def __len__(self):                     # runs on len(book)
        return self.pages

    def __eq__(self, other):               # runs on book1 == book2
        return self.title == other.title

b1 = Book("Python Basics", 300)
b2 = Book("Python Basics", 150)

print(b1)
print(len(b1))
print(b1 == b2)
````

## Key Points
- A **function** is standalone. A **method** is a function defined inside a class.
- A method needs `self` and is called through an object.
- A **special method** has double underscores on both sides and is called **automatically** by Python.
- You never write `b1.__str__()` or `b1.__len__()`. Use `print(b1)` and `len(b1)` instead.
- Without `__str__`, `print(b1)` shows something unhelpful like `<__main__.Book object at 0x...>`.
- Functions don't have to be global (you can nest one function inside another). The real difference is that a function is not inside a class and a method is.

# Topic 42 | Using Context Managers (with Statement)

## The problem
When we open a file, we must close it. If we forget, or if an error happens before `close()` runs, the file stays open. This wastes memory, can lock the file, and may lose unsaved data.

## Context manager
A context manager **sets something up** when a block starts and **cleans it up** when the block ends, automatically, even if an error happens.

We use it with the `with` statement:

    with open("demo.txt", "r") as file:
        data = file.read()

- `open(...)` creates the file object.
- `as file` gives it a name to use inside the block.
- When the block ends, the file is **closed automatically**. You never write `file.close()`.

````python
file = open("demo.txt", "w")
file.write("Hello\n")
file.close()                      # fine here, but we must remember it

# What if an error happens before close()?
file = open("demo.txt", "r")
try:
    data = file.read()
    result = 10 / 0               # error!
    file.close()                  # never reached
except ZeroDivisionError:
    print("Error happened")

print("closed?", file.closed)     # the file is still open
file.close()                      # manual cleanup
````

````python
with open("demo.txt", "w") as file:
    file.write("Hello, Python\n")
    file.write("Context managers are great\n")
    print("inside with, closed?", file.closed)

print("after with, closed?", file.closed)
````

````python
try:
    with open("demo.txt", "r") as file:
        print(file.read())
        result = 10 / 0           # error inside the block
except ZeroDivisionError:
    print("Error happened")

print("closed?", file.closed)     # still closed automatically
````

## Without `with` vs with `with`

| | `open()` + `close()` | `with open()` |
|---|---|---|
| Closes the file | Only if you remember | Always, automatically |
| If an error happens | The file may stay open | The file is still closed |
| Code length | Longer (needs `try/finally` to be safe) | Shorter and cleaner |
| Recommended | No | **Yes** |

## Notes
- The file object is only meant to be used **inside** the `with` block. After it, the file is closed, so reading it gives a `ValueError`.
- The variable (`file`) still exists after the block, but the file is closed.
- You can open several files in one `with` statement (next cell).

````python
with open("demo.txt", "r") as source, open("copy.txt", "w") as target:
    for line in source:
        target.write(line.upper())

with open("copy.txt") as file:
    print(file.read())
````

## How it works behind the scenes

Any object can work with `with` if its class has these 2 special methods (you learned special methods in Topic 39):

| Method | Runs | Purpose |
|---|---|---|
| `__enter__` | When the `with` block **starts** | Sets things up. What it returns goes into the `as` variable |
| `__exit__` | When the `with` block **ends**, even after an error | Cleans up (closes the file) |

Python's file object already has both, which is why `with open(...)` works.

In `__exit__(self, exc_type, exc_value, traceback)`, the 3 extra parameters describe an error if one happened. Returning `False` (or nothing) lets the error continue as normal. Returning `True` would hide it.

````python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode

    def __enter__(self):
        print("Opening file")
        self.file = open(self.filename, self.mode)
        return self.file                      # this goes into "as f"

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing file")
        self.file.close()
        return False                          # do not hide errors

with FileManager("demo.txt", "r") as f:
    print(f.read())

print(f.closed)
````

````python
from contextlib import contextmanager

@contextmanager
def open_file(name, mode):
    print("Opening file")
    f = open(name, mode)
    try:
        yield f                    # code before yield = __enter__
    finally:
        f.close()                  # code after yield = __exit__
        print("Closing file")

with open_file("demo.txt", "r") as f:
    print(f.readline(), end="")

print(f.closed)
````

````python
import time

class Timer:
    def __enter__(self):
        self.start = time.time()
        return self

    def __exit__(self, *args):
        print(f"Took {time.time() - self.start:.1f} seconds")

with Timer():
    time.sleep(0.5)
````

## Where context managers are used
- Files: `with open(...)` (the most common)
- Database connections: open, then always close
- Locks and threads: acquire, then always release
- Timers, temporary settings, test setups

## Key Points
- A context manager does **setup** at the start of a block and **cleanup** at the end.
- `with open(...) as file:` closes the file automatically, even if an error happens.
- Without `with`, you must call `close()` yourself, and an error can skip it.
- A class becomes a context manager by defining `__enter__` and `__exit__`.
- `@contextmanager` from `contextlib` is a shorter way to write one with a function and `yield`.
- Always use `with` for files. It is the standard, recommended way.

## Using Context Managers (other Statements)

`open()` is not the only context manager. Many other tools in Python's standard library work with `with` too, and they all give the same guarantee: **cleanup happens automatically**.

| Tool | What it does |
|---|---|
| `open()` | The standard file context manager for reading and writing files |
| `tempfile.TemporaryFile()` / `NamedTemporaryFile()` | Creates temporary files that **delete themselves** automatically |
| `zipfile.ZipFile()` | Opens `.zip` files for reading and writing |
| `contextlib.ExitStack` | Lets you manage **multiple files** at once easily |
| `pathlib.Path().open()` | Modern, object-oriented file opening through `pathlib` |

````python
import os
import tempfile

# TemporaryFile: no visible name, deleted when closed
with tempfile.TemporaryFile(mode="w+") as temp:
    temp.write("Temporary data")
    temp.seek(0)                      # go back to the start to read
    print(temp.read())

print("closed?", temp.closed)

# NamedTemporaryFile: has a name (temp.name) you can use
with tempfile.NamedTemporaryFile(mode="w+", suffix=".txt") as temp:
    temp.write("Hello from temp")
    temp.flush()                      # make sure the text is written
    name = temp.name
    print("inside, exists?", os.path.exists(name))

print("after, exists?", os.path.exists(name))
````

````python
import zipfile

# First create two small files to zip
with open("a.txt", "w") as f:
    f.write("File A\n")
with open("b.txt", "w") as f:
    f.write("File B\n")

# Write a zip file
with zipfile.ZipFile("archive.zip", "w") as z:
    z.write("a.txt")
    z.write("b.txt")

# Read a zip file
with zipfile.ZipFile("archive.zip", "r") as z:
    print(z.namelist())               # list of files inside
    print(z.read("a.txt").decode())   # read one file's content
````

````python
from contextlib import ExitStack

with open("c.txt", "w") as f:
    f.write("File C\n")

files = ["a.txt", "b.txt", "c.txt"]

with ExitStack() as stack:
    handles = [stack.enter_context(open(name)) for name in files]
    for h in handles:
        print(h.name, "->", h.readline().strip())

print([h.closed for h in handles])
````

````python
from pathlib import Path

p = Path("notes.txt")

with p.open("w") as f:                # same as open("notes.txt", "w")
    f.write("Written with pathlib\n")

with p.open("r") as f:
    print(f.read())

# One-line shortcuts (they open and close the file for you)
p.write_text("Short way\n")
print(p.read_text())

print(p.exists(), p.name, p.suffix)
````

## Summary: Which one to use?

| I want to... | Use |
|---|---|
| Read or write a normal file | `with open(...)` |
| Use a scratch file that cleans itself up | `with tempfile.TemporaryFile()` or `NamedTemporaryFile()` |
| Read or create a `.zip` archive | `with zipfile.ZipFile(...)` |
| Open a list of files, any number of them | `with ExitStack() as stack:` |
| Work with file paths in a modern way | `with Path(...).open(...)` |
| Quickly read or write a whole small file | `Path.read_text()` / `Path.write_text()` |

## Key Points
- Many Python tools are context managers, not just `open()`.
- They all follow the same idea: **set up at the start, clean up at the end**, even after an error.
- `tempfile` files delete themselves, so no leftover files.
- `ZipFile` closes the archive automatically.
- `ExitStack` manages a variable number of files with a single `with`.
- `pathlib` is the modern, readable way to work with paths and files.

# Topic 43 | Handling Exceptions During File Operations

## What is exception handling?
An **exception** is an error that happens while the program is running. Without handling, the program **crashes** at that line and the rest of the code never runs.

With exception handling, we use `try` and `except` to **catch** the error and respond to it, so the program **does not crash** and can continue.

Files are a common source of errors, because things outside our code can go wrong (the file is missing, we have no permission, and so on).

| Block | Purpose |
|---|---|
| `try` | Code that might cause an error |
| `except` | Runs **only if** that error happens |
| `else` | Runs only if **no** error happened |
| `finally` | **Always** runs, error or not |

````python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found. Please check the file name.")
except PermissionError:
    print("You don't have permission to open this file.")
else:
    print("File read successfully")
finally:
    print("Done (this always runs)")
````

````python
def read_file(filename):
    try:
        with open(filename, "r") as file:
            return file.read()
    except FileNotFoundError:
        print(f"'{filename}' does not exist.")
    except PermissionError:
        print(f"No permission to read '{filename}'.")
    except Exception as e:                 # catches any other error
        print("Unexpected error:", e)

content = read_file("missing.txt")
print("Program continues running")         # runs because the error was handled
````

## Common file exceptions

| Exception | When it happens |
|---|---|
| `FileNotFoundError` | The file doesn't exist (reading with `"r"`) |
| `PermissionError` | You don't have permission to open the file |
| `FileExistsError` | Using `"x"` mode on a file that already exists |
| `IsADirectoryError` | You tried to open a folder as a file |
| `UnicodeDecodeError` | The file content can't be decoded as text |

## Key Points
- No handling = the program **crashes**. With `try/except` = the error is caught and the program **continues**.
- Catch **specific** errors first (`FileNotFoundError`), and put the general `except Exception` last.
- Use `else` for code that should run only on success, and `finally` for cleanup that must always run.
- `with open(...)` already closes the file for you, so you usually don't need `finally` just to close it.
- Don't write a bare `except:` that hides every error. Name the error you expect.

# Topic 44 | Introduction to Exceptions in Python

## What is an exception?
An **exception** is an error that happens **while the program is running**, such as opening a file that doesn't exist (`FileNotFoundError`).

## Crash vs handled
- **Not handled:** Python stops the program at that line and shows an error message (a traceback). The rest of the code never runs.
- **Handled:** we tell Python in advance what to do if that error happens. The program does **not** crash and **continues** with the next lines.

We handle exceptions with the keywords `try` and `except`:
- `try`: the code that might cause an error
- `except`: what to do if that error happens

## Why handle exceptions?
- Prevents the program from crashing
- Lets us show a friendly message instead of a scary error
- Lets the program continue or exit in a controlled way

````python
file = open("missing.txt")
print("This line never runs")
````

````python
try:
    file = open("missing.txt")
    print(file.read())
except FileNotFoundError:
    print("File not found, but the program did not crash")

print("Program continues")
````

## Common exceptions

| Exception | Example | Cause |
|---|---|---|
| `FileNotFoundError` | `open("missing.txt")` | File doesn't exist |
| `ZeroDivisionError` | `10 / 0` | Dividing by zero |
| `ValueError` | `int("abc")` | Right type, wrong value |
| `TypeError` | `"5" + 5` | Wrong type for the operation |
| `IndexError` | `[1, 2, 3][5]` | List position doesn't exist |

| `NameError` | `print(undefined_var)` | Variable was never defined |

- An exception is a runtime error. If it is not handled, the program **crashes**.
- If it is handled with `try/except`, the program **continues**.

# Topic 45 | Using try, except, else & finally Blocks

## The four blocks

| Block | Runs when | Purpose |
|---|---|---|
| `try` | Always (it is the starting point) | Code that might cause an error |
| `except` | **Only if** an error happens in `try` | Handles the error |
| `else` | **Only if no** error happened in `try` | Code that should run on success |
| `finally` | **Always**, error or not | Cleanup that must always happen |

## How it flows
- **No error:** `try` -> `else` -> `finally`
- **Error:** `try` (stops at the error line) -> `except` -> `finally`

## Rules
- `try` needs at least one `except` or a `finally`.
- The order is fixed: `try`, `except`, `else`, `finally`.
- `else` and `finally` are optional.
- Put only the risky line(s) in `try`, and the success code in `else`. This way `except` catches only the error you expect.

````python
def divide(a, b):
    try:
        result = a / b                      # might fail
    except ZeroDivisionError:
        print("Cannot divide by zero")      # runs only on error
    else:
        print("Result:", result)            # runs only if no error
    finally:
        print("Done")                       # always runs

divide(10, 2)
print("---")
divide(10, 0)
````

````python
def read_file(name):
    try:
        file = open(name, "r")
    except FileNotFoundError:
        print("File not found")
    else:
        print(file.read())                  # only runs if the file opened
        file.close()
    finally:
        print("Finished")                   # always runs

read_file("demo.txt")        # a file that exists
print("---")
read_file("nope.txt")        # a file that doesn't
````

## Why use `else` instead of putting everything in `try`?
Code inside `try` is protected by `except`. If success code is also inside `try`, an unrelated error in it could be caught by mistake and hidden. With `else`, only the risky line is guarded, and an error in the `else` block is **not** caught by that same `try`'s `except`.

## Why use `finally`?
It runs **no matter what**: after success, after a handled error, and even if the `try` block uses `return`. It is the place for cleanup such as closing a file or a database connection.

## Key Points
- `try`: risky code. `except`: handle the error. `else`: runs on success. `finally`: always runs.
- Flow with no error: `try` -> `else` -> `finally`. Flow with an error: `try` -> `except` -> `finally`.
- Use `finally` for must-run cleanup. For files, `with open(...)` (Topic 42) already does this cleanup for you.
- Catch specific exceptions, not a bare `except:`.

# Topic 46 | Raising Exceptions in Python

## What does "raising" mean?
So far, Python raised the errors for us (e.g. `FileNotFoundError`). With the `raise` keyword, **we raise an exception ourselves** when our own rule is broken, even if Python itself sees nothing wrong.

Example: Python is happy with `age = -5`, but for our program a negative age makes no sense, so we raise an error.

## Syntax
    raise ExceptionType("message")

- `ExceptionType` is a built-in one (`ValueError`, `TypeError`, ...) or your own.
- The message explains **what went wrong**.
- When `raise` runs, the function stops at that line, exactly like a normal error. It crashes unless someone handles it with `try/except`.

## Why raise exceptions?
- Stop **bad data** early, before it causes bigger problems later.
- Tell the caller of a function that something is wrong, instead of failing silently.
- Make errors clear, with a message you wrote.

## Other forms
- `raise` alone, inside `except`: **re-raise** the same error after handling part of it (for example, after logging it).
- **Custom exception:** create your own error type by inheriting from `Exception` (you learned inheritance in Topic 36 onward).

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    print("Age set to", age)

set_age(20)

try:
    set_age(-5)
except ValueError as e:
    print("Caught:", e)
````

````python
class EmptyFileError(Exception):           # our own exception type
    pass

def read_notes(name):
    try:
        with open(name) as f:
            data = f.read()
    except FileNotFoundError:
        print("Missing file, passing the error up")
        raise                              # re-raise the same error

    if not data.strip():
        raise EmptyFileError(f"{name} is empty")
    return data

for n in ["full.txt", "empty.txt", "gone.txt"]:
    try:
        print(repr(read_notes(n)))
    except EmptyFileError as e:
        print("EmptyFileError:", e)
    except FileNotFoundError as e:
        print("FileNotFoundError:", e)
````

## Common exceptions to raise

| Exception | Use when |
|---|---|
| `ValueError` | The value has the right type but is not acceptable (negative age, empty name) |
| `TypeError` | The value has the wrong type (text where a number is needed) |
| `FileNotFoundError` | A required file is missing |
| `Exception` | A general error (better: create your own custom exception) |

## Raising vs handling

| | `raise` | `except` |
|---|---|---|
| Role | **Creates** the error | **Catches** the error |
| Used by | The code that finds the problem | The code that decides what to do about it |

## Key Points
- `raise ExceptionType("message")` creates an error on purpose when our own rule is broken.
- Raising stops the current function immediately. Handle it with `try/except` to keep the program running.
- Choose the exception type that matches the problem, and write a clear message.
- Bare `raise` inside `except` re-raises the same error.
- Create custom exceptions with `class MyError(Exception): pass`.

# Topic 47 | Creating Custom Exceptions in Python

## What is a custom exception?
Python has built-in exceptions (`ValueError`, `FileNotFoundError`, ...), but sometimes none of them describes **our own problem** clearly. A custom exception is an error type that **we create ourselves** for our program, such as `InvalidAgeError` or `InsufficientFundsError`.

## How to create one
A custom exception is a **class that inherits from `Exception`** (this is inheritance from Topic 36 onward):

    class InvalidAgeError(Exception):
        pass

Then we use it like any other exception: `raise` it, and catch it with `except`.

## Why create custom exceptions?
- **Clear names:** `InsufficientFundsError` says more than a generic `ValueError`.
- **Precise handling:** we can catch exactly our error and ignore the others.
- **Extra information:** we can attach data to the error, such as the balance and the amount.
- **Grouping:** related errors can share one parent class, so one `except` catches the whole group.

````python
class InvalidAgeError(Exception):
    pass

def set_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative")
    return age

try:
    set_age(-5)
except InvalidAgeError as e:
    print("Caught:", e)
````

````python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw Rs.{amount}, balance is only Rs.{balance}")

class Account:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance

acc = Account(1000)
print(acc.withdraw(300))                      # 700

try:
    acc.withdraw(5000)
except InsufficientFundsError as e:
    print("Caught:", e)
    print("Short by Rs.", e.amount - e.balance)

# Grouping related errors under one parent
class AppError(Exception): pass
class FileFormatError(AppError): pass
class EmptyFileError(AppError): pass

for exc in [FileFormatError("bad format"), EmptyFileError("empty")]:
    try:
        raise exc
    except AppError as e:                     # one except catches both children
        print(type(e).__name__, "caught by AppError:", e)
````

## Rules and good practice
- Inherit from `Exception`, **not** `BaseException`. `BaseException` includes system exits and keyboard interrupts, which you should not catch by accident.
- End the name with **`Error`** (`InvalidAgeError`). It is the Python convention.
- `pass` is enough for a simple one. Add `__init__` only when you want to store extra data.
- Use `super().__init__(message)` so the message works with `print(e)`.
- Make one **base** class for your project (`AppError`) and inherit the others from it. Callers can then catch one specific error or the whole group.
- Create a custom exception only when the built-in ones don't describe the problem well. For a simple bad value, `ValueError` is often enough.

## Key Points
- A custom exception is a **class that inherits from `Exception`**.
- Use `raise` to throw it and `except` to catch it, the same as built-in exceptions.
- It can carry **extra data** (attributes) through its own `__init__`.
- A parent exception class can catch all of its children with a single `except`.
- Clear names make errors easier to understand and debug.
