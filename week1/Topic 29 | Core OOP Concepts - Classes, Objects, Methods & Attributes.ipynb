## Classes and Objects

**Class = Blueprint:**
A class is a blueprint of the object. It defines what attributes 
and functions (methods) the object will have.

If we need a Car object, we first need a Car class that defines 
WHICH attributes and functions it needs.

**Class vs Object:**
- Class is ABSTRACT — it's just a definition/template, not a real thing yet.
- Object is a REAL INSTANCE — created from the class, it actually exists 
  in memory and can be used.

Using the `class` keyword tells Python: "I am creating a blueprint 
for an object." The class decides what the object will look like 
and how it will behave (its attributes and methods).

Once the class is created → THEN objects can be created from it.

**Methods inside a class:**
A class has its own functions, called METHODS (e.g. `__init__` and 
`start` in your example). These aren't global functions — they 
belong specifically to the class, and can also work with data 
(like `color`).

**The `self` keyword:**
`self` refers to the specific object the method is being called on. 
When you create an object and call a method on it, `self` represents 
THAT particular object — this is how each object keeps its own data 
separate, even though they share the same class blueprint.

## The `__init__` Method (Constructor)

**Definition:**
`__init__` is a special method in a class called a CONSTRUCTOR. 
It is called AUTOMATICALLY by Python the moment an object is 
created — you never call it yourself directly.

**Purpose:**
Its job is to set up (initialize) the object's initial attributes 
— i.e., assign starting values to the object's data as soon as 
it's created.

**How it's different from a regular method:**
- `__init__` → runs automatically, immediately when the object is created.
- A regular method (like `start()`) → only runs when you explicitly 
  call it on the object.

**Example:**
```python
class Car:
    def __init__(self, color):
        self.color = color
        print(f"A new {self.color} car object was created!")  # runs automatically

    def start(self):                                            # regular method
        print(f"{self.color} car started")

# __init__ runs automatically here, the moment the object is made
my_car = Car("red")
# Output: A new red car object was created!

# start() does NOT run automatically — only runs when explicitly called
my_car.start()
# Output: red car started
```

**Key takeaway:**
You will never see `my_car.__init__("red")` written in code — Python 
calls it for you behind the scenes as soon as you write `Car("red")`. 
Regular methods like `start()` always need to be called explicitly 
using dot notation.

````python
class Car:
    def __init__(self, color):
        self.color = color

    def start(self):
        print(f"{self.color} car started")

# First object
my_car = Car("red")
my_car.start()        # red car started

# Second object
mybrother_car = Car("red")
mybrother_car.start() # red car started
````
