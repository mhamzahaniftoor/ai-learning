# Topic 40 | Reading from & Writing to Files in Python

## Types of files
- **Text file:** plain, simple text (`.txt`)
- **CSV file:** comma-separated values, tabular data (`.csv`)
- **JSON file:** structured data in key-value form (`.json`)

## Why we need files
Variables disappear when the program ends. Files let us **store data permanently** and read it back later.

## The 3 steps
1. **Open** the file with `open()`
2. **Read or write** the data
3. **Close** the file with `close()`

Files are saved in the same folder as the notebook, unless you give a full path.

## File modes

| Mode | Meaning | If the file doesn't exist |
|---|---|---|
| `"r"` | Read (default) | Error: `FileNotFoundError` |
| `"w"` | Write, **erases old content** | Creates it |
| `"a"` | Append, adds to the end | Creates it |
| `"x"` | Create a new file only | Creates it (error if it already exists) |
| `"r+"` | Read and write | Error |
| `"b"` | Add for binary files, e.g. `"rb"` | Depends on the main mode |

```python
file = open("demo.txt", "w")
file.write("Hello, Python\n")
file.write("Another Line\n")
file.close()
```

```python
file = open("demo.txt", "r")
print(file.read())
file.close()
```

## Notes
- `\n` means a new line. `write()` does **not** add it for you.
- `write()` returns the number of characters written.
- **Always close the file.** Data may not be fully saved until `close()` is called, and an open file stays locked in memory.

```python
# readline() reads one line at a time
with open("demo.txt") as file:
    print(file.readline(), end="")
    print(file.readline(), end="")

# readlines() returns all lines as a list
with open("demo.txt") as file:
    lines = file.readlines()
print(lines)

# loop over the file (best for big files)
with open("demo.txt") as file:
    for line in file:
        print(line.strip())
```

## The with statement (best practice)
`with open(...) as file:` closes the file automatically, even if an error happens inside. You don't need to write `file.close()`.

| Method | Purpose |
|---|---|
| `read()` | Reads the whole file as one string |
| `readline()` | Reads one line |
| `readlines()` | Reads all lines into a list |
| `write(text)` | Writes a string |
| `writelines(list)` | Writes a list of strings |

```python
try:
    with open("missing.txt") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found")
```

```python
import csv, json

# CSV
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age"])
    writer.writerow(["Ali", 20])
    writer.writerow(["Sara", 16])

with open("students.csv") as file:
    for row in csv.reader(file):
        print(row)

# JSON
student = {"name": "Ali", "age": 20, "grade": "A"}
with open("student.json", "w") as file:
    json.dump(student, file)

with open("student.json") as file:
    data = json.load(file)
print(data["name"])
```

## Key Points
- Open with `open(name, mode)`, work with the file, then close it.
- Prefer `with open(...)`, because it closes the file automatically.
- `"w"` erases the old content, `"a"` adds to the end, `"r"` only reads.
- `write()` does not add a new line, so include `\n` yourself.
- Reading a missing file gives `FileNotFoundError`. Use `try/except` to handle it.
- Use the `csv` module for CSV files and the `json` module for JSON files.
