# Topic 12 | Python Data Types: Numbers, Strings & Booleans

A **data type** tells Python what kind of data a variable holds, and what operations are allowed on it.

## 1. Integer (`int`)
- Stores **whole numbers** (no decimal point): `19`, `-5`, `0`.
- Python integers have **no size limit**.

## 2. Float (`float`)
- Stores numbers **with a decimal point**: `3.0`, `3.5`, `-0.25`.
- `3` is an `int`, but `3.0` is a `float`.
- Division `/` always gives a float: `10 / 5` is `2.0`.

## 3. String (`str`)
- Stores **text**: letters, digits, spaces and special characters, written inside quotes (`"Ali"` or `'Ali'`).
- `"19"` in quotes is a string, not a number.
- Operations: `+` **concatenates** (joins), `*` **repeats**, `len()` gives the length, and `text[0]` gets one character.
- Strings are **immutable**: you can't change one character in place.
- `+` works only between two strings. `"Age: " + 19` gives a `TypeError`, so use `str(19)` or an f-string.

## 4. Boolean (`bool`)
- Stores only `True` or `False` (capital first letter).
- Often used as a **flag**: a switch that says yes or no, such as `is_valid = True`.
- Comes from comparisons: `5 > 3` is `True`.

```python
age = 19                 # int
gpa = 3.5                # float
is_valid = True          # bool

print(type(age), type(gpa), type(is_valid))

print(age + 1)           # 20
print(gpa * 2)           # 7.0
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)   # 3.5 3 1 8

print(10 / 5)            # 2.0 (division gives a float)
print(5 > 3, 5 == 3)     # True False

flag = False
if not flag:
    print("flag is off")
```

```python
name = "Ali"
father_name = "Hamza"

print(name + " " + father_name)     # Ali Hamza (concatenation)
print(name * 3)                     # AliAliAli (repetition)
print(len(name), name[0], name[-1], name.upper())   # 3 A i ALI

print("Age: " + str(19))            # convert the number to a string first
print(f"{name} is {age} years old") # f-string, the easiest way

print("19" + "1", 19 + 1)           # 191 20 (string vs number)

try:
    print("Age: " + 19)
except TypeError as e:
    print("TypeError:", e)
```

## Summary

| Type | Stores | Example | `type()` |
|---|---|---|---|
| Integer | Whole numbers | `19` | `int` |
| Float | Numbers with a decimal point | `3.5` | `float` |
| String | Text (letters, digits, spaces, symbols) | `"Ali"` | `str` |
| Boolean | `True` or `False` | `True` | `bool` |

## Key Points
- **Integer** = whole numbers. **Float** = decimal numbers. **String** = text in quotes. **Boolean** = `True` / `False`.
- `+` adds numbers but **joins** strings. You can't mix a string and a number with `+`, so convert with `str()` or use an f-string.
- A number in quotes (`"19"`) is a **string**.
- Floats can show tiny rounding errors: `0.1 + 0.2` gives `0.30000000000000004`. Use `round(x, 2)` when you need a clean result.
- `True` and `False` need a capital letter.
- Check any value's type with `type(value)`.
