## Getter and Setter

**Getter:**
A method used to GET (read) the value of a private attribute from
outside the class.
Example: get_grade()

**Setter:**
A method used to SET (change) the value of a private attribute from
outside the class.
Example: set_grade()

**Why we need them:**
- A private attribute (like __grade) cannot be accessed directly
  from outside the class.
- Getter and setter are the controlled doors to that data.
- The setter can validate the new value before accepting it, so
  invalid data never gets in. Direct access cannot do this.

**Naming convention:**
get_attributeName() and set_attributeName()

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade              # private attribute

    # Getter: reads the private attribute
    def get_grade(self):
        return self.__grade

    # Setter: changes the private attribute, with validation
    def set_grade(self, new_grade):
        if new_grade in ["A", "B", "C", "D", "F"]:
            self.__grade = new_grade
        else:
            print("Invalid grade")

student1 = Student("Ali", "A")

print(student1.get_grade())   # A      (using the getter)

student1.set_grade("B")       # using the setter
print(student1.get_grade())   # B

student1.set_grade("Z")       # Invalid grade (setter rejected it)
print(student1.get_grade())   # B      (value unchanged)
```
