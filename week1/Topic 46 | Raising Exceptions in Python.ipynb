# Topic 46 | Raising Exceptions in Python

## What does "raising" mean?
So far, Python raised the errors for us (e.g. `FileNotFoundError`). With the `raise` keyword, **we raise an exception ourselves** when our own rule is broken, even if Python itself sees nothing wrong.

Example: Python is happy with `age = -5`, but for our program a negative age makes no sense, so we raise an error.

## Syntax
    raise ExceptionType("message")

- `ExceptionType` is a built-in one (`ValueError`, `TypeError`, ...) or your own.
- The message explains **what went wrong**.
- When `raise` runs, the function stops at that line, exactly like a normal error. It crashes unless someone handles it with `try/except`.

## Why raise exceptions?
- Stop **bad data** early, before it causes bigger problems later.
- Tell the caller of a function that something is wrong, instead of failing silently.
- Make errors clear, with a message you wrote.

## Other forms
- `raise` alone, inside `except`: **re-raise** the same error after handling part of it (for example, after logging it).
- **Custom exception:** create your own error type by inheriting from `Exception` (you learned inheritance in Topic 36 onward).

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    print("Age set to", age)

set_age(20)

try:
    set_age(-5)
except ValueError as e:
    print("Caught:", e)
```

```python
class EmptyFileError(Exception):           # our own exception type
    pass

def read_notes(name):
    try:
        with open(name) as f:
            data = f.read()
    except FileNotFoundError:
        print("Missing file, passing the error up")
        raise                              # re-raise the same error

    if not data.strip():
        raise EmptyFileError(f"{name} is empty")
    return data

for n in ["full.txt", "empty.txt", "gone.txt"]:
    try:
        print(repr(read_notes(n)))
    except EmptyFileError as e:
        print("EmptyFileError:", e)
    except FileNotFoundError as e:
        print("FileNotFoundError:", e)
```

## Common exceptions to raise

| Exception | Use when |
|---|---|
| `ValueError` | The value has the right type but is not acceptable (negative age, empty name) |
| `TypeError` | The value has the wrong type (text where a number is needed) |
| `FileNotFoundError` | A required file is missing |
| `Exception` | A general error (better: create your own custom exception) |

## Raising vs handling

| | `raise` | `except` |
|---|---|---|
| Role | **Creates** the error | **Catches** the error |
| Used by | The code that finds the problem | The code that decides what to do about it |

## Key Points
- `raise ExceptionType("message")` creates an error on purpose when our own rule is broken.
- Raising stops the current function immediately. Handle it with `try/except` to keep the program running.
- Choose the exception type that matches the problem, and write a clear message.
- Bare `raise` inside `except` re-raises the same error.
- Create custom exceptions with `class MyError(Exception): pass`.
