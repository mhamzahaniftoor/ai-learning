## Variable Scope: Local vs Global

**Local variable:**
A variable declared inside a function is called a local variable.
It only exists while the function is running, and can't be accessed outside it.

**Global variable:**
A variable declared outside of any function (at the program/module level) is called a global variable.

**Global variables — accessible everywhere (but risky):**
A global variable is accessible to every function in the program.
This is not a good approach — it can cause confusion during debugging, 
since it may get mixed up with local variables of the same name.

**Best practice:**
Prefer using local variables over global ones — it keeps functions 
self-contained and easier to debug.

```python
x = 10   # global variable

def my_func():
    y = 5   # local variable
    print(x)   # can access global x
    print(y)   # can access local y

my_func()
print(x)   # works fine — x is global
print(y)   # ERROR — y only exists inside my_func()
```

```python
def show():
    x=5 //local
    return x
```
