## Encapsulation and Private Attributes

**Encapsulation:**
Keeping an object's data (attributes) and functions (methods) together
inside the object, and exposing only what is needed. Like a mouse, we use
the left and right buttons (the functionality) without seeing the circuit
(the core logic).

**Private attributes:**
- A private attribute is meant to be used only inside the class, not
  from outside it.
- We make an attribute private by putting two underscores before its
  name: `__grade`.
- Outside code cannot read it directly (`student1.__grade` gives an error).
- We access it through a method instead. A method that reads the value
  is called a GETTER.

**Why use it:**
- Protects data from accidental or invalid changes.
- Lets the class control how its data is read and modified.
- Internal logic can change without breaking the code that uses the class.

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade        # private attribute

    def get_grade(self):            # getter
        return self.__grade

student1 = Student("Ali", "A")

print(student1.name)         # Ali (public, direct access works)
print(student1.get_grade())  # A (private, accessed through the getter)
print(student1.__grade)      # AttributeError: no attribute '__grade'
```

Setter: the natural partner of a getter. It lets you change a private value with validation, and this is where encapsulation pays off:

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.__grade = grade

    def get_grade(self):
        return self.__grade

    def set_grade(self, new_grade):
        if new_grade in ["A", "B", "C", "D", "F"]:
            self.__grade = new_grade
        else:
            print("Invalid grade")

student1 = Student("Ali", "A")
student1.set_grade("B")
print(student1.get_grade())   # B
student1.set_grade("Z")       # Invalid grade
```
