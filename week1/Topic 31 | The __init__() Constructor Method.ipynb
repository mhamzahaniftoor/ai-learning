## Constructor (`__init__`) — Recap

This method is automatically called when the object is created — 
it is called the CONSTRUCTOR METHOD, and it sets the initial 
values for the object's attributes.

```python
class Car:
    def __init__(self, color="white"):   # "white" is a DEFAULT value
        self.color = color                # this assigns the INITIAL value

my_car1 = Car("red")     # color = "red"   → initial value you gave
my_car2 = Car()           # color = "white" → default value used, since none was given
```
