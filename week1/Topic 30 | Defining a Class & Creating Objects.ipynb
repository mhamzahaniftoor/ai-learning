```python
class Car:
    def __init__(self,color,model):
        self.color=color
        self.model=model

    def start(self):
        print(f"My car is {self.color}")
```

```python
car = Car("Black", "Civic")
```

```python
car.start()
```

```python
class Car:
    color = "Red"
    def drive(self):
        print(f"{self.color} car is moving")

    def setColor(self,color):
        self.color = color
        print(f"{self.color} car is moving")
```

```python
car = Car()
car.drive()
```
