# Topic 45 | Using try, except, else & finally Blocks

## The four blocks

| Block | Runs when | Purpose |
|---|---|---|
| `try` | Always (it is the starting point) | Code that might cause an error |
| `except` | **Only if** an error happens in `try` | Handles the error |
| `else` | **Only if no** error happened in `try` | Code that should run on success |
| `finally` | **Always**, error or not | Cleanup that must always happen |

## How it flows
- **No error:** `try` -> `else` -> `finally`
- **Error:** `try` (stops at the error line) -> `except` -> `finally`

## Rules
- `try` needs at least one `except` or a `finally`.
- The order is fixed: `try`, `except`, `else`, `finally`.
- `else` and `finally` are optional.
- Put only the risky line(s) in `try`, and the success code in `else`. This way `except` catches only the error you expect.

```python
def divide(a, b):
    try:
        result = a / b                      # might fail
    except ZeroDivisionError:
        print("Cannot divide by zero")      # runs only on error
    else:
        print("Result:", result)            # runs only if no error
    finally:
        print("Done")                       # always runs

divide(10, 2)
print("---")
divide(10, 0)
```

```python
def read_file(name):
    try:
        file = open(name, "r")
    except FileNotFoundError:
        print("File not found")
    else:
        print(file.read())                  # only runs if the file opened
        file.close()
    finally:
        print("Finished")                   # always runs

read_file("demo.txt")        # a file that exists
print("---")
read_file("nope.txt")        # a file that doesn't
```

## Why use `else` instead of putting everything in `try`?
Code inside `try` is protected by `except`. If success code is also inside `try`, an unrelated error in it could be caught by mistake and hidden. With `else`, only the risky line is guarded, and an error in the `else` block is **not** caught by that same `try`'s `except`.

## Why use `finally`?
It runs **no matter what**: after success, after a handled error, and even if the `try` block uses `return`. It is the place for cleanup such as closing a file or a database connection.

## Key Points
- `try`: risky code. `except`: handle the error. `else`: runs on success. `finally`: always runs.
- Flow with no error: `try` -> `else` -> `finally`. Flow with an error: `try` -> `except` -> `finally`.
- Use `finally` for must-run cleanup. For files, `with open(...)` (Topic 42) already does this cleanup for you.
- Catch specific exceptions, not a bare `except:`.
