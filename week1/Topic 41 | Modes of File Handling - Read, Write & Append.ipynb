# Topic 41 | Modes of File Handling: Read, Write & Append

The **mode** is the second argument of `open(filename, mode)`. It tells Python **what you plan to do** with the file and **how to treat the existing content**.

## The three main modes

| Mode | Name | What it does | If the file doesn't exist | Old content |
|---|---|---|---|---|
| `"r"` | Read (default) | Read only | Error: `FileNotFoundError` | Kept |
| `"w"` | Write | Write only | Creates it | **Erased** |
| `"a"` | Append | Write only, at the end | Creates it | Kept |

```python
try:
    file = open("demo.txt", "r")      # "r" is the default mode
except FileNotFoundError:
    print("File not found. Read mode never creates a file.")
```

```python
# 1) "w" creates the file
with open("demo.txt", "w") as file:
    file.write("Line 1\n")
    file.write("Line 2\n")

with open("demo.txt", "r") as file:
    print(file.read())

# 2) "w" again ERASES everything and starts fresh
with open("demo.txt", "w") as file:
    file.write("Fresh start\n")

with open("demo.txt", "r") as file:
    print(file.read())
```

```python
with open("demo.txt", "a") as file:
    file.write("Added 1\n")
    file.write("Added 2\n")

with open("demo.txt", "r") as file:
    print(file.read())
```

## Write vs Append (the most common mistake)

- `"w"` **erases** the old content the moment you open the file, even if you write nothing.
- `"a"` **keeps** the old content and adds new text at the end.
- If you want to add to a file, use `"a"`. Using `"w"` by mistake **destroys your data**.
- Both modes **create** the file if it doesn't exist.
- `write()` never adds a new line for you, so include `\n` yourself.
