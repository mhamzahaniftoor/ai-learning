# Topic 43 | Handling Exceptions During File Operations

## What is exception handling?
An **exception** is an error that happens while the program is running. Without handling, the program **crashes** at that line and the rest of the code never runs.

With exception handling, we use `try` and `except` to **catch** the error and respond to it, so the program **does not crash** and can continue.

Files are a common source of errors, because things outside our code can go wrong (the file is missing, we have no permission, and so on).

| Block | Purpose |
|---|---|
| `try` | Code that might cause an error |
| `except` | Runs **only if** that error happens |
| `else` | Runs only if **no** error happened |
| `finally` | **Always** runs, error or not |

```python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found. Please check the file name.")
except PermissionError:
    print("You don't have permission to open this file.")
else:
    print("File read successfully")
finally:
    print("Done (this always runs)")
```

```python
def read_file(filename):
    try:
        with open(filename, "r") as file:
            return file.read()
    except FileNotFoundError:
        print(f"'{filename}' does not exist.")
    except PermissionError:
        print(f"No permission to read '{filename}'.")
    except Exception as e:                 # catches any other error
        print("Unexpected error:", e)

content = read_file("missing.txt")
print("Program continues running")         # runs because the error was handled
```

## Common file exceptions

| Exception | When it happens |
|---|---|
| `FileNotFoundError` | The file doesn't exist (reading with `"r"`) |
| `PermissionError` | You don't have permission to open the file |
| `FileExistsError` | Using `"x"` mode on a file that already exists |
| `IsADirectoryError` | You tried to open a folder as a file |
| `UnicodeDecodeError` | The file content can't be decoded as text |

## Key Points
- No handling = the program **crashes**. With `try/except` = the error is caught and the program **continues**.
- Catch **specific** errors first (`FileNotFoundError`), and put the general `except Exception` last.
- Use `else` for code that should run only on success, and `finally` for cleanup that must always run.
- `with open(...)` already closes the file for you, so you usually don't need `finally` just to close it.
- Don't write a bare `except:` that hides every error. Name the error you expect.
