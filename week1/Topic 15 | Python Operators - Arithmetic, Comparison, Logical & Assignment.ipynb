# Topic 15 | Python Operators: Arithmetic, Comparison, Logical & Assignment

An **operator** is a symbol that performs an operation on values. The values it works on are called **operands**. In `2 + 3`, the operator is `+` and the operands are `2` and `3`.

## By number of operands
- **Unary** operator: **one** operand. Examples: `-5` (negative), `+5`, `not True`.
- **Binary** operator: **two** operands. Examples: `2 + 3`, `5 > 2`, `a and b`.

## By purpose: 4 groups

### 1. Arithmetic (maths)
| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `2 + 3` | `5` |
| `-` | Subtraction | `2 - 3` | `-1` |
| `*` | Multiplication | `2 * 6` | `12` |
| `/` | Division (always a float) | `2 / 9` | `0.2222...` |
| `//` | Floor division (rounds **down**) | `2 // 9` | `0` |
| `%` | Modulus (remainder) | `101 % 10` | `1` |
| `**` | Power | `2 ** 4` | `16` |

`%` is handy for even/odd checks (`n % 2 == 0`), and `//` with `%` splits a number into quotient and remainder. Floor division rounds toward **negative** infinity: `-7 // 2` is `-4`, not `-3`.

### 2. Comparison (the result is always `True` or `False`)
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `2 == 4` | `False` |
| `!=` | Not equal to | `2 != 2` | `False` |
| `>` | Greater than | `2 > 6` | `False` |
| `<` | Less than | `2 < 1` | `False` |
| `>=` | Greater than or equal | `2 >= 2` | `True` |
| `<=` | Less than or equal | `2 <= 1` | `False` |

Python allows **chained** comparisons: `1 < age < 100`.

### 3. Logical (combine True/False values)
| Operator | Result is `True` when | Type |
|---|---|---|
| `and` | **both** sides are `True` | binary |
| `or` | **at least one** side is `True` | binary |
| `not` | reverses the value (`not True` is `False`) | unary |

`and` / `or` use **short-circuit** evaluation: Python stops as soon as the answer is known. They return one of the operands, not always `True`/`False`: `0 or "default"` gives `"default"`.

### 4. Assignment (store or update a value)
| Operator | Same as |
|---|---|
| `=` | store a value: `x = 10` |
| `+=` | `x = x + 5` |
| `-=` | `x = x - 3` |
| `*=` | `x = x * 2` |
| `/=` | `x = x / 4` |
| `//=` | `x = x // 5` |
| `%=` | `x = x % 2` |
| `**=` | `x = x ** 3` |

```python
# Arithmetic (binary: two operands)
print(2 + 3)             # 5
print(2 - 3)             # -1
print(2 * 6)             # 12
print(2 / 9)             # 0.2222222222222222 (always a float)
print(2 // 9)            # 0 (floor division)
print(101 % 10)          # 1 (remainder)
print(2 ** 4)            # 16 (power)
print(-7 // 2, -7 % 2)   # -4 1 (floor division rounds DOWN)

# Unary (one operand)
x = 5
print(-x, +x, not True)  # -5 5 False

# Comparison (result is always True or False)
print(2 == 4)            # False
print(2 != 2)            # False
print(2 > 6)             # False
print(2 < 1)             # False
print(2 >= 2, 2 <= 1)    # True False

age = 25
print(1 < age < 100)     # True (chained comparison)
print("apple" == "apple", "a" < "b", 2 == 2.0)   # True True True
```

```python
age = 25
has_id = True

# Logical
print(age >= 18 and has_id)      # True  (both True)
print(age < 18 or has_id)        # True  (one is True)
print(not has_id)                # False (reversed)
print(0 or "default", 5 and 6)   # default 6 (they return an operand)

# Assignment operators
x = 10
x += 5; print(x)     # 15
x -= 3; print(x)     # 12
x *= 2; print(x)     # 24
x /= 4; print(x)     # 6.0
y = 17
y //= 5; print(y)    # 3
y %= 2; print(y)     # 1
y = 3
y **= 3; print(y)    # 27

# Precedence: which operator runs first
print(2 + 3 * 4)     # 14 (multiplication first)
print((2 + 3) * 4)   # 20 (brackets first)
print(2 ** 3 ** 2)   # 512 (power goes right to left: 2 ** 9)
print(-2 ** 2)       # -4 (power before the minus)

# A common trap with floats
print(0.1 + 0.2 == 0.3)               # False
print(round(0.1 + 0.2, 2) == 0.3)     # True
```

## Operator precedence (highest to lowest)

| Order | Operators |
|---|---|
| 1 | `( )` brackets |
| 2 | `**` |
| 3 | unary `+x`, `-x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | comparisons: `==`, `!=`, `>`, `<`, `>=`, `<=` |
| 7 | `not` |
| 8 | `and` |
| 9 | `or` |

When unsure, **use brackets**. They make the order clear to the reader.

## Key Points
- **Unary** = one operand (`-5`, `not x`). **Binary** = two operands (`2 + 3`).
- `/` always returns a float. `//` rounds down. `%` gives the remainder. `**` is power.
- Comparison operators always return `True` or `False`. Don't confuse `=` (assign) with `==` (compare).
- Logical operators are the words `and`, `or`, `not`.
- `x += 1` is shorthand for `x = x + 1`. Python has no `x++`.
- Floats can be slightly inexact (`0.1 + 0.2 != 0.3`), so use `round()` when comparing them.
