## Functions in Python

**Definition:**
A function is a block of code that performs a specific task.

**Benefits of using functions:**
1. Reusability We can reuse this block of code multiple times without rewriting it.
2. Organization  Helps make the code organized, using meaningful naming conventions for functions.
    
**Important Note:**
A function does not execute itself  it only runs when you call it.

```python
def sum(num1,num2):
    return num1+num2
```

```python
sum(5,4)
```
