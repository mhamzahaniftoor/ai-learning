## Method Overriding

**Definition:**
Method overriding is when a child class defines a method with the
same name as a method in its parent class, so the child's version
replaces (overrides) the parent's version for the child's objects.

**Why use it:**
- The parent gives a general behaviour (e.g. every Animal can speak).
- Each child can change it to its own specific behaviour
  (a Dog barks, a Cat meows) without touching the parent.

**Rules:**
- The method name must be exactly the same in parent and child.
- Python looks in the child class first. If it finds the method there,
  it uses it and never goes to the parent.
- If the child does not define the method, the parent's version is used.

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):                # overrides Animal.speak
        print("Woof woof")

class Cat(Animal):
    def speak(self):                # overrides Animal.speak
        print("Meow")

class Cow(Animal):
    pass                            # no override, uses the parent's version

animal1 = Animal()
dog1 = Dog()
cat1 = Cat()
cow1 = Cow()

animal1.speak()   # Animal makes a sound
dog1.speak()      # Woof woof
cat1.speak()      # Meow
cow1.speak()      # Animal makes a sound (inherited, not overridden)
```

```python
class Dog(Animal):
    def speak(self):
        super().speak()             # run the parent's version first
        print("Woof woof")          # then add Dog's own behaviour

Dog().speak()
# Animal makes a sound
# Woof woof
```
