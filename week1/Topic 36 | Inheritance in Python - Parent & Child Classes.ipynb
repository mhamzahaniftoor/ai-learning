## Inheritance

**Definition:**
Inheritance lets one class (the child) automatically get the attributes
and methods of another class (the parent). It is an advanced form of
code reusability.

**Terms:**
- Parent class (also called base class or superclass): the class being inherited from, e.g. Animal
- Child class (also called derived class or subclass): the class that inherits, e.g. Dog, Cat

**How it works:**
- Write the shared behaviour once in the parent (e.g. speak() in Animal).
- Each child class gets it for free, so you don't rewrite it.
- Each child can also add its own unique behaviour (Dog has bark(), Cat has meow()).

**Syntax:**
class Child(Parent):
The parent's name goes in brackets after the child's name.

```python
class Animal:                       # parent class
    def speak(self):
        print("Speak")

class Dog(Animal):                  # child class
    def bark(self):
        print("Woof!")

class Cat(Animal):                  # another child class
    def meow(self):
        print("Meow!")

dog1 = Dog()
dog1.bark()     # Woof!   (Dog's own method)
dog1.speak()    # Speak   (inherited from Animal)

cat1 = Cat()
cat1.meow()     # Meow!   (Cat's own method)
cat1.speak()    # Speak   (inherited from Animal)

# dog1.meow()   # AttributeError: Dog doesn't have meow
```

Python looks in the child class first, and only goes to the parent if it doesn't find the method.

Inheritance with __init__ and super(): if the child has its own __init__, it should call the parent's __init__ using super(), otherwise the parent's attributes never get set:

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)      # runs Animal's __init__ for name
        self.breed = breed          # Dog's own attribute

    def bark(self):
        print(f"{self.name} says Woof!")

dog1 = Dog("Tommy", "Labrador")
dog1.speak()    # Tommy makes a sound
dog1.bark()     # Tommy says Woof!
print(dog1.breed)   # Labrador
```
